## 1. Introduction: What Are Parametric Tests?

**Parametric tests** are a class of statistical hypothesis tests that make explicit assumptions about the **parameters** of the population distribution from which the data are drawn. The term "parametric" derives from the fact that these tests estimate and test hypotheses about population **parameters** (e.g., mean $$\mu$$, variance $$\sigma^2$$, proportion $$p$$).

### Key Characteristics:

| Feature | Description |
| --- | --- |
| **Distribution Assumption** | Data follows a known probability distribution (typically Normal/Gaussian) |
| **Scale of Measurement** | Requires interval or ratio-scale data |
| **Parameters Tested** | Population mean, variance, correlation, regression coefficients |
| **Statistical Power** | Higher power than non-parametric counterparts when assumptions are met |
| **Sample Size** | Can work with smaller samples (due to distributional assumptions) |

### The General Framework of Hypothesis Testing:

Every parametric test follows this structure:

1. **State Hypotheses**: Define $$H_0$$ (null) and $$H_1$$ (alternative)
2. **Choose Significance Level**: Set $$\alpha$$ (typically 0.05)
3. **Compute Test Statistic**: Calculate a value that summarizes the evidence against $$H_0$$
4. **Determine Critical Region or p-value**: Compare test statistic to its sampling distribution
5. **Make Decision**: Reject or fail to reject $$H_0$$

## 2. Parametric vs. Non-Parametric Tests

| Aspect | Parametric | Non-Parametric |
| --- | --- | --- |
| **Assumptions** | Normality, homogeneity of variance, interval data | Distribution-free, works with ordinal data |
| **Central Tendency** | Tests means | Tests medians or ranks |
| **Power** | Higher (when assumptions met) | Lower (but robust to violations) |
| **Sample Size** | Effective with small $$n$$ | Often needs larger $$n$$ |
| **Examples** | t-test, ANOVA, Pearson $$r$$ | Mann-Whitney, Kruskal-Wallis, Spearman $$\rho$$ |

**Rule of Thumb**: Use parametric tests when:
- Data is continuous (interval/ratio scale)
- Distribution is approximately normal (or $$n > 30$$ by CLT)
- Variances are roughly equal across groups
- No extreme outliers contaminate the data

## 3. Core Assumptions of Parametric Tests

All parametric tests rest on some combination of these assumptions:

### 3.1 Normality

The data (or residuals) are drawn from a normally distributed population:

$$f(x) = \frac{1}{\sigma\sqrt{2\pi}} \exp\left(-\frac{(x - \mu)^2}{2\sigma^2}\right)$$

**How to check:**
- Visual: Q-Q plot, histogram
- Formal tests: Shapiro-Wilk test, Kolmogorov-Smirnov test, Anderson-Darling test
- Skewness $$\approx 0$$, Kurtosis $$\approx 3$$ (excess kurtosis $$\approx 0$$)

**Robustness**: The t-test and ANOVA are fairly robust to moderate violations of normality when $$n \geq 30$$ (Central Limit Theorem).

### 3.2 Homogeneity of Variance (Homoscedasticity)

Population variances across groups are equal:

$$\sigma_1^2 = \sigma_2^2 = \cdots = \sigma_k^2$$

**How to check:**
- Levene's test (robust to non-normality)
- Bartlett's test (sensitive to non-normality)
- Brown-Forsythe test

**When violated**: Use Welch's t-test (does not assume equal variances) or Welch's ANOVA.

### 3.3 Independence

Observations are statistically independent of one another:

$$P(X_i, X_j) = P(X_i) \cdot P(X_j) \quad \forall \; i \neq j$$

**Ensured by**: Random sampling, proper experimental design. Violations occur with repeated measures, clustered data, or time-series autocorrelation.

### 3.4 Scale of Measurement

Data must be measured on an **interval** or **ratio** scale (i.e., meaningful numerical differences and/or ratios). Ordinal or nominal data require non-parametric alternatives.

### 3.5 Random Sampling

The sample should be a random draw from the population of interest, ensuring that the sample statistics are unbiased estimators of population parameters.

## 4. One-Sample Z-Test

### Purpose
Test whether the population mean $$\mu$$ differs from a hypothesized value $$\mu_0$$ when the population standard deviation $$\sigma$$ is **known**.

### Hypotheses

| Type | $$H_0$$ | $$H_1$$ |
| --- | --- | --- |
| Two-tailed | $$\mu = \mu_0$$ | $$\mu \neq \mu_0$$ |
| Left-tailed | $$\mu = \mu_0$$ | $$\mu < \mu_0$$ |
| Right-tailed | $$\mu = \mu_0$$ | $$\mu > \mu_0$$ |

### Test Statistic

$$Z = \frac{\bar{X} - \mu_0}{\sigma / \sqrt{n}}$$

Where:
- $$\bar{X}$$ = sample mean
- $$\mu_0$$ = hypothesized population mean
- $$\sigma$$ = known population standard deviation
- $$n$$ = sample size

Under $$H_0$$, $$Z \sim N(0, 1)$$ (standard normal distribution).

### Decision Rule (at significance level $$\alpha$$)

- **Two-tailed**: Reject $$H_0$$ if $$|Z| > Z_{\alpha/2}$$
- **Left-tailed**: Reject $$H_0$$ if $$Z < -Z_{\alpha}$$
- **Right-tailed**: Reject $$H_0$$ if $$Z > Z_{\alpha}$$

### Confidence Interval

$$\bar{X} \pm Z_{\alpha/2} \cdot \frac{\sigma}{\sqrt{n}}$$

### Assumptions
1. Population standard deviation $$\sigma$$ is known
2. Data is continuous (interval/ratio)
3. Sample is randomly drawn
4. Population is normally distributed (or $$n \geq 30$$ for CLT)

### Industrial Applications
- **Manufacturing**: Testing if a production line's output mean weight equals the specification (e.g., 500g cereal boxes)
- **Pharmaceuticals**: Verifying that the mean active ingredient concentration matches the labeled value
- **Quality Control**: Checking if the mean diameter of machined parts meets engineering tolerances
- **Finance**: Testing if the average return of a portfolio differs from a benchmark

````python
# =============================================================================
# ONE-SAMPLE Z-TEST: Manufacturing Quality Control Example
# =============================================================================
# Scenario: A cereal factory claims boxes contain 500g on average.
# Population std dev (from historical data) is known: sigma = 15g
# We sample 36 boxes and test if the mean differs from 500g.

# Generate sample data
np.random.seed(42)
n = 36
mu_0 = 500           # Hypothesized population mean
sigma = 15           # Known population std dev
sample = np.random.normal(loc=495, scale=15, size=n)  # True mean is 495

# Calculate test statistic manually
x_bar = sample.mean()
z_stat = (x_bar - mu_0) / (sigma / np.sqrt(n))
p_value_two_tailed = 2 * (1 - stats.norm.cdf(abs(z_stat)))

print("="*60)
print("ONE-SAMPLE Z-TEST RESULTS")
print("="*60)
print(f"\nSample size (n):              {n}")
print(f"Sample mean (x̄):             {x_bar:.4f}")
print(f"Hypothesized mean (μ₀):       {mu_0}")
print(f"Known population σ:           {sigma}")
print(f"\nZ-statistic:                  {z_stat:.4f}")
print(f"p-value (two-tailed):         {p_value_two_tailed:.6f}")
print(f"Critical values (α=0.05):     ±{stats.norm.ppf(0.975):.4f}")

# Using statsmodels ztest
z_stat_sm, p_val_sm = ztest(sample, value=mu_0)
print(f"\n--- Verification with statsmodels ---")
print(f"Z-statistic:                  {z_stat_sm:.4f}")
print(f"p-value:                      {p_val_sm:.6f}")

# Confidence Interval
alpha = 0.05
z_crit = stats.norm.ppf(1 - alpha/2)
ci_lower = x_bar - z_crit * (sigma / np.sqrt(n))
ci_upper = x_bar + z_crit * (sigma / np.sqrt(n))
print(f"\n95% Confidence Interval:      ({ci_lower:.4f}, {ci_upper:.4f})")

# Interpretation
print(f"\n{'='*60}")
print("INTERPRETATION:")
if p_value_two_tailed < alpha:
    print(f"  Since p-value ({p_value_two_tailed:.6f}) < α ({alpha}),")
    print(f"  we REJECT H₀. There is significant evidence that the")
    print(f"  population mean differs from {mu_0}g.")
else:
    print(f"  Since p-value ({p_value_two_tailed:.6f}) ≥ α ({alpha}),")
    print(f"  we FAIL TO REJECT H₀. Insufficient evidence to conclude")
    print(f"  the mean differs from {mu_0}g.")
print("="*60)

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Sample distribution
axes[0].hist(sample, bins=12, edgecolor='black', alpha=0.7, density=True)
axes[0].axvline(x_bar, color='red', linestyle='--', linewidth=2, label=f'Sample Mean = {x_bar:.2f}')
axes[0].axvline(mu_0, color='green', linestyle='-', linewidth=2, label=f'H₀: μ = {mu_0}')
axes[0].set_xlabel('Weight (g)')
axes[0].set_ylabel('Density')
axes[0].set_title('Sample Distribution of Box Weights')
axes[0].legend()

# Plot 2: Z-test on standard normal
x_range = np.linspace(-4, 4, 1000)
axes[1].plot(x_range, stats.norm.pdf(x_range), 'b-', linewidth=2)
axes[1].fill_between(x_range[x_range <= -z_crit], stats.norm.pdf(x_range[x_range <= -z_crit]), alpha=0.3, color='red', label='Rejection Region')
axes[1].fill_between(x_range[x_range >= z_crit], stats.norm.pdf(x_range[x_range >= z_crit]), alpha=0.3, color='red')
axes[1].axvline(z_stat, color='darkred', linestyle='--', linewidth=2, label=f'Z = {z_stat:.3f}')
axes[1].set_xlabel('Z-value')
axes[1].set_ylabel('Density')
axes[1].set_title('Standard Normal Distribution with Rejection Regions')
axes[1].legend()

plt.tight_layout()
plt.show()
````

## 5. One-Sample t-Test

### Purpose
Test whether the population mean $$\mu$$ differs from a hypothesized value $$\mu_0$$ when the population standard deviation $$\sigma$$ is **unknown** (estimated from the sample).

### Why Not Z-Test?
In practice, $$\sigma$$ is almost never known. William Sealy Gosset (publishing as "Student" in 1908) developed the t-distribution to account for the additional uncertainty from estimating $$\sigma$$ with the sample standard deviation $$s$$.

### Test Statistic

$$t = \frac{\bar{X} - \mu_0}{s / \sqrt{n}}$$

Where:
- $$s = \sqrt{\frac{1}{n-1}\sum_{i=1}^{n}(X_i - \bar{X})^2}$$ (sample standard deviation)

Under $$H_0$$, $$t \sim t_{n-1}$$ (Student's t-distribution with $$n-1$$ degrees of freedom).

### The t-Distribution

The PDF of the t-distribution with $$\nu$$ degrees of freedom:

$$f(t) = \frac{\Gamma\left(\frac{\nu+1}{2}\right)}{\sqrt{\nu\pi}\;\Gamma\left(\frac{\nu}{2}\right)} \left(1 + \frac{t^2}{\nu}\right)^{-\frac{\nu+1}{2}}$$

**Properties:**
- Symmetric, bell-shaped, centered at 0
- Heavier tails than the standard normal (more conservative)
- As $$\nu \to \infty$$, $$t_\nu \to N(0,1)$$
- For $$\nu > 30$$, the t-distribution is very close to the standard normal

### Effect Size (Cohen's d)

$$d = \frac{\bar{X} - \mu_0}{s}$$

| $$|d|$$ | Interpretation |
| --- | --- |
| 0.2 | Small effect |
| 0.5 | Medium effect |
| 0.8 | Large effect |

### Assumptions
1. Population $$\sigma$$ is unknown
2. Data is continuous
3. Random sampling
4. Approximate normality (critical for small $$n$$, less so for large $$n$$)

### Industrial Applications
- **Clinical Trials**: Testing if a new drug's mean recovery time differs from the standard treatment
- **A/B Testing (small samples)**: Comparing a new website feature's mean engagement time to baseline
- **Education**: Testing if a class's average test score differs from the national average
- **Environmental Science**: Checking if a river's mean pollutant level exceeds safety standards

````python
# =============================================================================
# ONE-SAMPLE t-TEST: Clinical Trial Example
# =============================================================================
# Scenario: A pharmaceutical company claims their new pain reliever
# reduces headache duration to an average of 30 minutes.
# We test 25 patients and measure their headache duration after taking the drug.

np.random.seed(123)
n = 25
mu_0 = 30  # Claimed mean duration
sample_t = np.random.normal(loc=27, scale=8, size=n)  # True mean = 27 min

# Manual calculation
x_bar_t = sample_t.mean()
s = sample_t.std(ddof=1)  # Sample std dev (Bessel's correction)
t_stat = (x_bar_t - mu_0) / (s / np.sqrt(n))
df = n - 1
p_value_t = 2 * stats.t.cdf(-abs(t_stat), df=df)  # Two-tailed

print("="*60)
print("ONE-SAMPLE t-TEST RESULTS")
print("="*60)
print(f"\nSample size (n):              {n}")
print(f"Sample mean (x̄):             {x_bar_t:.4f}")
print(f"Sample std dev (s):           {s:.4f}")
print(f"Hypothesized mean (μ₀):       {mu_0}")
print(f"Degrees of freedom (df):      {df}")
print(f"\nt-statistic:                  {t_stat:.4f}")
print(f"p-value (two-tailed):         {p_value_t:.6f}")
print(f"Critical values (α=0.05):     ±{stats.t.ppf(0.975, df=df):.4f}")

# Using scipy
t_stat_scipy, p_val_scipy = stats.ttest_1samp(sample_t, popmean=mu_0)
print(f"\n--- Verification with scipy ---")
print(f"t-statistic:                  {t_stat_scipy:.4f}")
print(f"p-value:                      {p_val_scipy:.6f}")

# Effect Size (Cohen's d)
cohens_d = (x_bar_t - mu_0) / s
print(f"\nEffect Size (Cohen's d):      {cohens_d:.4f}")
if abs(cohens_d) < 0.2:
    print("  → Negligible effect")
elif abs(cohens_d) < 0.5:
    print("  → Small effect")
elif abs(cohens_d) < 0.8:
    print("  → Medium effect")
else:
    print("  → Large effect")

# Confidence Interval
t_crit = stats.t.ppf(0.975, df=df)
ci_lower_t = x_bar_t - t_crit * (s / np.sqrt(n))
ci_upper_t = x_bar_t + t_crit * (s / np.sqrt(n))
print(f"\n95% Confidence Interval:      ({ci_lower_t:.4f}, {ci_upper_t:.4f})")

alpha = 0.05
print(f"\n{'='*60}")
print("INTERPRETATION:")
if p_val_scipy < alpha:
    print(f"  Since p-value ({p_val_scipy:.6f}) < α ({alpha}),")
    print(f"  we REJECT H₀. Significant evidence that mean headache")
    print(f"  duration ≠ {mu_0} minutes.")
    print(f"  The drug appears to reduce duration below the claimed {mu_0} min.")
else:
    print(f"  Since p-value ({p_val_scipy:.6f}) ≥ α ({alpha}),")
    print(f"  we FAIL TO REJECT H₀.")
print("="*60)

# Visualization: t-distribution
fig, ax = plt.subplots(figsize=(10, 5))
x_range = np.linspace(-5, 5, 1000)
ax.plot(x_range, stats.t.pdf(x_range, df=df), 'b-', linewidth=2, label=f't-distribution (df={df})')
ax.plot(x_range, stats.norm.pdf(x_range), 'g--', linewidth=1.5, label='Standard Normal', alpha=0.7)
ax.fill_between(x_range[x_range <= -t_crit], stats.t.pdf(x_range[x_range <= -t_crit], df=df), alpha=0.3, color='red', label='Rejection Region (α=0.05)')
ax.fill_between(x_range[x_range >= t_crit], stats.t.pdf(x_range[x_range >= t_crit], df=df), alpha=0.3, color='red')
ax.axvline(t_stat, color='darkred', linestyle='--', linewidth=2, label=f't-stat = {t_stat:.3f}')
ax.set_xlabel('t-value')
ax.set_ylabel('Density')
ax.set_title('One-Sample t-Test: t-Distribution vs Standard Normal')
ax.legend()
plt.tight_layout()
plt.show()
````

## 6. Independent Two-Sample t-Test

### Purpose
Compare the means of **two independent groups** to determine if they come from populations with equal means.

### Hypotheses

$$H_0: \mu_1 = \mu_2 \quad (\text{or equivalently } \mu_1 - \mu_2 = 0)$$
$$H_1: \mu_1 \neq \mu_2 \quad (\text{two-tailed})$$

### Test Statistic

#### Case 1: Equal Variances Assumed (Student's t-test)

$$t = \frac{\bar{X}_1 - \bar{X}_2}{s_p \sqrt{\frac{1}{n_1} + \frac{1}{n_2}}}$$

Where $$s_p$$ is the **pooled standard deviation**:

$$s_p = \sqrt{\frac{(n_1 - 1)s_1^2 + (n_2 - 1)s_2^2}{n_1 + n_2 - 2}}$$

Degrees of freedom: $$df = n_1 + n_2 - 2$$

#### Case 2: Unequal Variances (Welch's t-test)

$$t = \frac{\bar{X}_1 - \bar{X}_2}{\sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}}$$

Degrees of freedom (Welch-Satterthwaite approximation):

$$df = \frac{\left(\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}\right)^2}{\frac{\left(\frac{s_1^2}{n_1}\right)^2}{n_1 - 1} + \frac{\left(\frac{s_2^2}{n_2}\right)^2}{n_2 - 1}}$$

### Effect Size (Cohen's d for independent samples)

$$d = \frac{\bar{X}_1 - \bar{X}_2}{s_p}$$

### Assumptions
1. Two samples are **independent** (no pairing/matching)
2. Data in each group is approximately normally distributed
3. Homogeneity of variance (for Student's version; Welch's relaxes this)
4. Continuous data on interval/ratio scale

### Industrial Applications
- **A/B Testing**: Comparing conversion rates between control and treatment groups
- **Manufacturing**: Comparing tensile strength of materials from two suppliers
- **HR Analytics**: Comparing average productivity between remote and in-office workers
- **Healthcare**: Comparing blood pressure reduction between two medications
- **Agriculture**: Comparing crop yields between two fertilizer treatments

````python
# =============================================================================
# INDEPENDENT TWO-SAMPLE t-TEST: A/B Testing Example
# =============================================================================
# Scenario: An e-commerce company runs an A/B test on their checkout page.
# Group A (control): original design
# Group B (treatment): new streamlined design
# Metric: Time spent on checkout (seconds) - lower is better

np.random.seed(42)

# Generate data
group_a = np.random.normal(loc=180, scale=45, size=50)  # Control: mean 180s
group_b = np.random.normal(loc=155, scale=40, size=55)  # Treatment: mean 155s

# Descriptive statistics
print("="*60)
print("INDEPENDENT TWO-SAMPLE t-TEST: A/B Test Results")
print("="*60)
print(f"\n{'Statistic':<25} {'Group A (Control)':<20} {'Group B (Treatment)'}")
print(f"{'-'*65}")
print(f"{'n':<25} {len(group_a):<20} {len(group_b)}")
print(f"{'Mean':<25} {group_a.mean():<20.4f} {group_b.mean():.4f}")
print(f"{'Std Dev':<25} {group_a.std(ddof=1):<20.4f} {group_b.std(ddof=1):.4f}")
print(f"{'Variance':<25} {group_a.var(ddof=1):<20.4f} {group_b.var(ddof=1):.4f}")

# Test for equal variances (Levene's test)
lev_stat, lev_p = stats.levene(group_a, group_b)
print(f"\n--- Levene's Test for Equal Variances ---")
print(f"F-statistic: {lev_stat:.4f}, p-value: {lev_p:.4f}")
equal_var = lev_p > 0.05
print(f"Equal variances assumed: {equal_var}")

# Student's t-test (equal variances)
t_student, p_student = stats.ttest_ind(group_a, group_b, equal_var=True)
print(f"\n--- Student's t-test (Equal Variances) ---")
print(f"t-statistic: {t_student:.4f}")
print(f"p-value: {p_student:.6f}")

# Welch's t-test (unequal variances)
t_welch, p_welch = stats.ttest_ind(group_a, group_b, equal_var=False)
print(f"\n--- Welch's t-test (Unequal Variances) ---")
print(f"t-statistic: {t_welch:.4f}")
print(f"p-value: {p_welch:.6f}")

# Effect Size
s_pooled = np.sqrt(((len(group_a)-1)*group_a.var(ddof=1) + (len(group_b)-1)*group_b.var(ddof=1)) / (len(group_a)+len(group_b)-2))
cohens_d_2 = (group_a.mean() - group_b.mean()) / s_pooled
print(f"\nPooled Std Dev: {s_pooled:.4f}")
print(f"Cohen's d: {cohens_d_2:.4f}")

alpha = 0.05
print(f"\n{'='*60}")
print("INTERPRETATION:")
print(f"  Mean difference: {group_a.mean() - group_b.mean():.2f} seconds")
if p_welch < alpha:
    print(f"  The new checkout design significantly reduces time spent.")
    print(f"  Effect size (d={cohens_d_2:.2f}) indicates a {'small' if abs(cohens_d_2)<0.5 else 'medium' if abs(cohens_d_2)<0.8 else 'large'} effect.")
    print(f"  RECOMMENDATION: Deploy the new design.")
else:
    print(f"  No statistically significant difference between designs.")
print("="*60)

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Box plot comparison
data_df = pd.DataFrame({'Time (seconds)': np.concatenate([group_a, group_b]),
                        'Group': ['A (Control)']*len(group_a) + ['B (Treatment)']*len(group_b)})
sns.boxplot(data=data_df, x='Group', y='Time (seconds)', ax=axes[0], palette='Set2')
axes[0].set_title('Checkout Time: Control vs Treatment')
axes[0].axhline(y=group_a.mean(), color='blue', linestyle='--', alpha=0.5, label=f'Mean A = {group_a.mean():.1f}')
axes[0].axhline(y=group_b.mean(), color='orange', linestyle='--', alpha=0.5, label=f'Mean B = {group_b.mean():.1f}')
axes[0].legend()

# Overlapping distributions
axes[1].hist(group_a, bins=15, alpha=0.5, label=f'Group A (μ̂={group_a.mean():.1f})', density=True, edgecolor='black')
axes[1].hist(group_b, bins=15, alpha=0.5, label=f'Group B (μ̂={group_b.mean():.1f})', density=True, edgecolor='black')
axes[1].set_xlabel('Time (seconds)')
axes[1].set_ylabel('Density')
axes[1].set_title('Distribution Overlap')
axes[1].legend()

plt.tight_layout()
plt.show()
````

## 7. Paired t-Test (Dependent Samples t-Test)

### Purpose
Compare means of **two related measurements** on the same subjects (before/after, matched pairs, repeated measures).

### Key Insight
Instead of comparing two groups directly, we analyze the **differences** $$D_i = X_{1i} - X_{2i}$$ and test whether the mean difference is zero.

### Hypotheses

$$H_0: \mu_D = 0 \quad (\text{no mean difference})$$
$$H_1: \mu_D \neq 0 \quad (\text{two-tailed})$$

### Test Statistic

$$t = \frac{\bar{D}}{s_D / \sqrt{n}}$$

Where:
- $$\bar{D} = \frac{1}{n}\sum_{i=1}^{n} D_i$$ = mean of the differences
- $$s_D = \sqrt{\frac{1}{n-1}\sum_{i=1}^{n}(D_i - \bar{D})^2}$$ = std dev of differences
- $$n$$ = number of pairs

Degrees of freedom: $$df = n - 1$$

### Why Paired > Independent?

The paired design **controls for individual variability**. By computing within-subject differences, we remove between-subject variance, yielding:
- Smaller standard error
- More statistical power
- More precise estimates

### Assumptions
1. Pairs are **related** (same subject, matched subjects)
2. Differences $$D_i$$ are approximately normally distributed
3. Differences are independent of each other
4. Continuous data

### Industrial Applications
- **Clinical Trials**: Pre-treatment vs. post-treatment measurements on the same patients
- **Education**: Comparing student scores before and after a training intervention
- **UX Research**: User satisfaction ratings before and after a UI redesign (same users)
- **Sports Science**: Athlete performance before and after a training program
- **Manufacturing**: Measuring part dimensions before and after a calibration process

````python
# =============================================================================
# PAIRED t-TEST: Blood Pressure Medication Study
# =============================================================================
# Scenario: A clinical trial measures systolic blood pressure (mmHg)
# of 30 patients before and after 8 weeks of a new medication.

np.random.seed(55)
n_pairs = 30

# Generate correlated before/after data
bp_before = np.random.normal(loc=145, scale=12, size=n_pairs)
# After treatment: mean reduction of ~10 mmHg with individual variation
bp_after = bp_before - np.random.normal(loc=10, scale=5, size=n_pairs)

# Compute differences
differences = bp_before - bp_after

# Manual calculation
d_bar = differences.mean()
s_d = differences.std(ddof=1)
t_paired = d_bar / (s_d / np.sqrt(n_pairs))
df_paired = n_pairs - 1
p_paired = 2 * (1 - stats.t.cdf(abs(t_paired), df=df_paired))

print("="*60)
print("PAIRED t-TEST RESULTS: Blood Pressure Study")
print("="*60)
print(f"\n{'Measure':<30} {'Before':<15} {'After':<15} {'Difference'}")
print(f"{'-'*70}")
print(f"{'Mean':<30} {bp_before.mean():<15.2f} {bp_after.mean():<15.2f} {d_bar:.2f}")
print(f"{'Std Dev':<30} {bp_before.std(ddof=1):<15.2f} {bp_after.std(ddof=1):<15.2f} {s_d:.2f}")
print(f"{'Min':<30} {bp_before.min():<15.2f} {bp_after.min():<15.2f} {differences.min():.2f}")
print(f"{'Max':<30} {bp_before.max():<15.2f} {bp_after.max():<15.2f} {differences.max():.2f}")

print(f"\n--- Test Results ---")
print(f"Mean difference (D̄):         {d_bar:.4f} mmHg")
print(f"Std dev of differences (s_D): {s_d:.4f}")
print(f"t-statistic:                  {t_paired:.4f}")
print(f"Degrees of freedom:           {df_paired}")
print(f"p-value (two-tailed):         {p_paired:.8f}")

# Using scipy
t_scipy, p_scipy = stats.ttest_rel(bp_before, bp_after)
print(f"\n--- scipy verification ---")
print(f"t-statistic: {t_scipy:.4f}, p-value: {p_scipy:.8f}")

# Effect Size
cohens_d_paired = d_bar / s_d
print(f"\nCohen's d (paired): {cohens_d_paired:.4f} (→ {'Large' if abs(cohens_d_paired) >= 0.8 else 'Medium' if abs(cohens_d_paired) >= 0.5 else 'Small'} effect)")

# CI for mean difference
t_crit_p = stats.t.ppf(0.975, df=df_paired)
ci_low_d = d_bar - t_crit_p * (s_d / np.sqrt(n_pairs))
ci_high_d = d_bar + t_crit_p * (s_d / np.sqrt(n_pairs))
print(f"95% CI for mean difference:   ({ci_low_d:.2f}, {ci_high_d:.2f}) mmHg")

alpha = 0.05
print(f"\n{'='*60}")
print("INTERPRETATION:")
print(f"  The medication reduced BP by an average of {d_bar:.1f} mmHg.")
print(f"  This reduction is statistically significant (p = {p_scipy:.2e}).")
print(f"  We are 95% confident the true reduction is between")
print(f"  {ci_low_d:.1f} and {ci_high_d:.1f} mmHg.")
print("="*60)

# Visualization
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# Before vs After paired lines
for i in range(n_pairs):
    axes[0].plot([0, 1], [bp_before[i], bp_after[i]], 'o-', alpha=0.4, color='gray')
axes[0].plot([0, 1], [bp_before.mean(), bp_after.mean()], 'o-', color='red', linewidth=3, markersize=10, label='Mean')
axes[0].set_xticks([0, 1])
axes[0].set_xticklabels(['Before', 'After'])
axes[0].set_ylabel('Systolic BP (mmHg)')
axes[0].set_title('Individual Patient Trajectories')
axes[0].legend()

# Distribution of differences
axes[1].hist(differences, bins=10, edgecolor='black', alpha=0.7, density=True)
axes[1].axvline(0, color='red', linestyle='-', linewidth=2, label='H₀: μ_D = 0')
axes[1].axvline(d_bar, color='green', linestyle='--', linewidth=2, label=f'D̄ = {d_bar:.1f}')
axes[1].set_xlabel('Difference (Before - After)')
axes[1].set_ylabel('Density')
axes[1].set_title('Distribution of Differences')
axes[1].legend()

# Q-Q plot to check normality of differences
stats.probplot(differences, dist="norm", plot=axes[2])
axes[2].set_title('Q-Q Plot of Differences (Normality Check)')

plt.tight_layout()
plt.show()
````

## 8. One-Way ANOVA (Analysis of Variance)

### Purpose
Compare the means of **three or more independent groups** simultaneously. ANOVA tests whether at least one group mean is significantly different from the others.

### Why Not Multiple t-Tests?
With $$k$$ groups, there are $$\binom{k}{2}$$ pairwise comparisons. Each t-test at $$\alpha = 0.05$$ inflates the **familywise error rate**:

$$\alpha_{\text{family}} = 1 - (1 - \alpha)^m$$

For $$k = 4$$ groups: $$m = 6$$ comparisons → $$\alpha_{\text{family}} = 1 - 0.95^6 = 0.265$$ (26.5% Type I error!)

ANOVA maintains the overall $$\alpha$$ at 0.05 for a single omnibus test.

### Hypotheses

$$H_0: \mu_1 = \mu_2 = \cdots = \mu_k \quad (\text{all group means are equal})$$
$$H_1: \text{At least one } \mu_i \text{ differs}$$

### The Logic of ANOVA: Partitioning Variance

Total variability is decomposed into:

$$SS_{\text{Total}} = SS_{\text{Between}} + SS_{\text{Within}}$$

Where:

$$SS_{\text{Total}} = \sum_{i=1}^{k}\sum_{j=1}^{n_i}(X_{ij} - \bar{X}_{\cdot\cdot})^2$$

$$SS_{\text{Between}} = \sum_{i=1}^{k} n_i(\bar{X}_{i\cdot} - \bar{X}_{\cdot\cdot})^2$$

$$SS_{\text{Within}} = \sum_{i=1}^{k}\sum_{j=1}^{n_i}(X_{ij} - \bar{X}_{i\cdot})^2$$

### Mean Squares

$$MS_{\text{Between}} = \frac{SS_{\text{Between}}}{k - 1}$$

$$MS_{\text{Within}} = \frac{SS_{\text{Within}}}{N - k}$$

Where $$N = \sum n_i$$ (total sample size).

### Test Statistic (F-ratio)

$$F = \frac{MS_{\text{Between}}}{MS_{\text{Within}}} \sim F_{k-1, \; N-k}$$

Under $$H_0$$, both mean squares estimate $$\sigma^2$$, so $$F \approx 1$$.
If group means truly differ, $$MS_{\text{Between}}$$ becomes large, pushing $$F$$ >> 1.

### Effect Size (Eta-squared)

$$\eta^2 = \frac{SS_{\text{Between}}}{SS_{\text{Total}}}$$

Interpretation: Proportion of total variance explained by group membership.

| $$\eta^2$$ | Interpretation |
| --- | --- |
| 0.01 | Small |
| 0.06 | Medium |
| 0.14 | Large |

### Post-Hoc Tests (When $$H_0$$ is Rejected)

ANOVA only tells us *something* differs. Post-hoc tests identify *which* pairs:

| Test | When to Use |
| --- | --- |
| **Tukey's HSD** | Equal sample sizes, all pairwise comparisons |
| **Bonferroni** | Conservative, any comparison set |
| **Scheffé** | Most conservative, complex contrasts |
| **Dunnett** | Comparing all groups to a single control |
| **Games-Howell** | Unequal variances, unequal sample sizes |

### Assumptions
1. **Independence**: Observations within and between groups are independent
2. **Normality**: Each group's data is approximately normally distributed
3. **Homogeneity of Variance**: All groups have equal population variances ($$\sigma_1^2 = \sigma_2^2 = \cdots = \sigma_k^2$$)

### Industrial Applications
- **Pharmaceuticals**: Comparing efficacy of 3+ drug doses
- **Agriculture**: Comparing crop yields across 4+ fertilizer types
- **Marketing**: Comparing click-through rates across multiple ad creatives
- **Manufacturing**: Comparing defect rates across different production shifts
- **Education**: Comparing test scores across multiple teaching methods

````python
# =============================================================================
# ONE-WAY ANOVA: Comparing Teaching Methods
# =============================================================================
# Scenario: A university tests 4 teaching methods on student exam performance.
# Method A: Traditional lecture
# Method B: Flipped classroom
# Method C: Online self-paced
# Method D: Hybrid (lecture + online)

np.random.seed(42)

# Generate data for 4 groups
method_a = np.random.normal(loc=72, scale=10, size=30)  # Traditional
method_b = np.random.normal(loc=78, scale=9, size=30)   # Flipped
method_c = np.random.normal(loc=70, scale=11, size=30)  # Online
method_d = np.random.normal(loc=80, scale=10, size=30)  # Hybrid

# Descriptive statistics
print("="*70)
print("ONE-WAY ANOVA: Comparing Teaching Methods")
print("="*70)
print(f"\n{'Method':<20} {'n':<6} {'Mean':<10} {'Std Dev':<10} {'Min':<10} {'Max'}")
print("-"*70)
for name, data in [('A (Traditional)', method_a), ('B (Flipped)', method_b), 
                    ('C (Online)', method_c), ('D (Hybrid)', method_d)]:
    print(f"{name:<20} {len(data):<6} {data.mean():<10.2f} {data.std(ddof=1):<10.2f} {data.min():<10.2f} {data.max():.2f}")

# Check assumptions
print(f"\n--- Assumption Checks ---")

# Normality (Shapiro-Wilk)
print("\nShapiro-Wilk Normality Test:")
for name, data in [('A', method_a), ('B', method_b), ('C', method_c), ('D', method_d)]:
    stat, p = stats.shapiro(data)
    print(f"  Method {name}: W={stat:.4f}, p={p:.4f} {'\u2713 Normal' if p > 0.05 else '\u2717 Non-normal'}")

# Homogeneity of variance (Levene's test)
lev_stat, lev_p = stats.levene(method_a, method_b, method_c, method_d)
print(f"\nLevene's Test: F={lev_stat:.4f}, p={lev_p:.4f} {'\u2713 Equal variances' if lev_p > 0.05 else '\u2717 Unequal variances'}")

# Perform One-Way ANOVA
f_stat, p_anova = stats.f_oneway(method_a, method_b, method_c, method_d)

# Manual ANOVA table
all_data = np.concatenate([method_a, method_b, method_c, method_d])
grand_mean = all_data.mean()
k = 4
N = len(all_data)

ss_between = sum(len(g) * (g.mean() - grand_mean)**2 for g in [method_a, method_b, method_c, method_d])
ss_within = sum(((g - g.mean())**2).sum() for g in [method_a, method_b, method_c, method_d])
ss_total = ((all_data - grand_mean)**2).sum()

ms_between = ss_between / (k - 1)
ms_within = ss_within / (N - k)
f_manual = ms_between / ms_within

print(f"\n{'='*70}")
print("ANOVA TABLE")
print(f"{'='*70}")
print(f"{'Source':<15} {'SS':<12} {'df':<6} {'MS':<12} {'F':<10} {'p-value'}")
print(f"{'-'*70}")
print(f"{'Between':<15} {ss_between:<12.4f} {k-1:<6} {ms_between:<12.4f} {f_manual:<10.4f} {p_anova:.6f}")
print(f"{'Within':<15} {ss_within:<12.4f} {N-k:<6} {ms_within:<12.4f}")
print(f"{'Total':<15} {ss_total:<12.4f} {N-1:<6}")

# Effect size
eta_squared = ss_between / ss_total
omega_squared = (ss_between - (k-1)*ms_within) / (ss_total + ms_within)
print(f"\nη² (eta-squared): {eta_squared:.4f} ({eta_squared*100:.1f}% variance explained)")
print(f"ω² (omega-squared): {omega_squared:.4f} (less biased estimate)")

alpha = 0.05
print(f"\n{'='*70}")
print("INTERPRETATION:")
if p_anova < alpha:
    print(f"  F({k-1}, {N-k}) = {f_stat:.2f}, p = {p_anova:.6f}")
    print(f"  REJECT H₀: At least one teaching method produces significantly")
    print(f"  different exam scores.")
    print(f"  \n  Proceeding with Tukey's HSD post-hoc test...")
print("="*70)

# Post-hoc: Tukey's HSD
from statsmodels.stats.multicomp import pairwise_tukeyhsd

groups = (['A']*30 + ['B']*30 + ['C']*30 + ['D']*30)
tukey = pairwise_tukeyhsd(all_data, groups, alpha=0.05)
print(f"\n{tukey}")

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Box plot
df_anova = pd.DataFrame({'Score': all_data, 'Method': groups})
sns.boxplot(data=df_anova, x='Method', y='Score', ax=axes[0], palette='Set2')
axes[0].axhline(grand_mean, color='red', linestyle='--', alpha=0.7, label=f'Grand Mean = {grand_mean:.1f}')
axes[0].set_title('Exam Scores by Teaching Method')
axes[0].set_xlabel('Teaching Method')
axes[0].set_ylabel('Exam Score')
axes[0].legend()

# Mean plot with CI
means = [method_a.mean(), method_b.mean(), method_c.mean(), method_d.mean()]
sems = [stats.sem(method_a), stats.sem(method_b), stats.sem(method_c), stats.sem(method_d)]
axes[1].errorbar(['A', 'B', 'C', 'D'], means, yerr=[1.96*s for s in sems], 
                 fmt='o-', capsize=8, capthick=2, linewidth=2, markersize=10)
axes[1].set_xlabel('Teaching Method')
axes[1].set_ylabel('Mean Exam Score')
axes[1].set_title('Mean Scores with 95% Confidence Intervals')
axes[1].axhline(grand_mean, color='red', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
````

## 9. Two-Way ANOVA (Factorial ANOVA)

### Purpose
Examine the effect of **two independent categorical factors** on a continuous dependent variable, including their **interaction effect**.

### Model

$$X_{ijk} = \mu + \alpha_i + \beta_j + (\alpha\beta)_{ij} + \epsilon_{ijk}$$

Where:
- $$\mu$$ = grand mean
- $$\alpha_i$$ = main effect of Factor A (level $$i$$)
- $$\beta_j$$ = main effect of Factor B (level $$j$$)
- $$(\alpha\beta)_{ij}$$ = interaction effect of A and B
- $$\epsilon_{ijk} \sim N(0, \sigma^2)$$ = random error

### Hypotheses (Three Separate Tests)

1. **Main Effect of Factor A**: $$H_0: \alpha_1 = \alpha_2 = \cdots = \alpha_a = 0$$
2. **Main Effect of Factor B**: $$H_0: \beta_1 = \beta_2 = \cdots = \beta_b = 0$$
3. **Interaction Effect**: $$H_0: (\alpha\beta)_{ij} = 0 \; \forall \; i, j$$

### Sum of Squares Decomposition

$$SS_{\text{Total}} = SS_A + SS_B + SS_{AB} + SS_{\text{Error}}$$

### ANOVA Table

| Source | SS | df | MS | F |
| --- | --- | --- | --- | --- |
| Factor A | $$SS_A$$ | $$a - 1$$ | $$MS_A = \frac{SS_A}{a-1}$$ | $$\frac{MS_A}{MS_E}$$ |
| Factor B | $$SS_B$$ | $$b - 1$$ | $$MS_B = \frac{SS_B}{b-1}$$ | $$\frac{MS_B}{MS_E}$$ |
| A × B | $$SS_{AB}$$ | $$(a-1)(b-1)$$ | $$MS_{AB} = \frac{SS_{AB}}{(a-1)(b-1)}$$ | $$\frac{MS_{AB}}{MS_E}$$ |
| Error | $$SS_E$$ | $$N - ab$$ | $$MS_E = \frac{SS_E}{N-ab}$$ | |
| Total | $$SS_T$$ | $$N - 1$$ | | |

### Interpreting Interactions

- **No interaction**: Effects of Factor A are consistent across levels of Factor B (parallel lines in interaction plot)
- **Interaction present**: The effect of one factor depends on the level of the other factor (non-parallel/crossing lines)

### Assumptions
1. Independence of observations
2. Normality within each cell (combination of factor levels)
3. Homogeneity of variance across all cells
4. Balanced design preferred (equal $$n$$ in each cell)

### Industrial Applications
- **Agriculture**: Effect of fertilizer type AND irrigation method on crop yield
- **Marketing**: Effect of price level AND advertising channel on sales
- **Manufacturing**: Effect of temperature AND pressure on product quality
- **Psychology**: Effect of therapy type AND session frequency on outcomes
- **Pharmaceuticals**: Effect of drug dosage AND patient age group on response

````python
# =============================================================================
# TWO-WAY ANOVA: Manufacturing Process Optimization
# =============================================================================
# Scenario: A factory tests how Temperature (Low, Medium, High) and
# Pressure (Low, High) affect product tensile strength (MPa).

np.random.seed(42)

# Create factorial design data
temperatures = ['Low', 'Medium', 'High']
pressures = ['Low', 'High']
n_per_cell = 15

data_rows = []
for temp in temperatures:
    for pressure in pressures:
        # Base strength + effects
        base = 50
        temp_effect = {'Low': -5, 'Medium': 0, 'High': 8}[temp]
        pressure_effect = {'Low': -3, 'High': 4}[pressure]
        # Interaction: High temp + High pressure gives extra boost
        interaction = 5 if (temp == 'High' and pressure == 'High') else 0
        
        values = np.random.normal(loc=base + temp_effect + pressure_effect + interaction, 
                                  scale=4, size=n_per_cell)
        for v in values:
            data_rows.append({'Temperature': temp, 'Pressure': pressure, 'Strength': v})

df_2way = pd.DataFrame(data_rows)

print("="*70)
print("TWO-WAY ANOVA: Temperature × Pressure on Tensile Strength")
print("="*70)

# Cell means
cell_means = df_2way.groupby(['Temperature', 'Pressure'])['Strength'].agg(['mean', 'std', 'count'])
print(f"\nCell Means:")
print(cell_means.round(2))
print(f"\nMarginal Means by Temperature:")
print(df_2way.groupby('Temperature')['Strength'].mean().round(2))
print(f"\nMarginal Means by Pressure:")
print(df_2way.groupby('Pressure')['Strength'].mean().round(2))

# Fit Two-Way ANOVA using statsmodels
model = ols('Strength ~ C(Temperature) * C(Pressure)', data=df_2way).fit()
anova_table = anova_lm(model, typ=2)

print(f"\n{'='*70}")
print("TWO-WAY ANOVA TABLE (Type II SS)")
print(f"{'='*70}")
print(anova_table.round(4))

# Effect sizes (partial eta-squared)
print(f"\n--- Effect Sizes (Partial η²) ---")
ss_residual = anova_table.loc['Residual', 'sum_sq']
for source in anova_table.index[:-1]:
    partial_eta = anova_table.loc[source, 'sum_sq'] / (anova_table.loc[source, 'sum_sq'] + ss_residual)
    print(f"  {source}: η²_p = {partial_eta:.4f}")

alpha = 0.05
print(f"\n{'='*70}")
print("INTERPRETATION:")
for source in anova_table.index[:-1]:
    p_val = anova_table.loc[source, 'PR(>F)']
    sig = "SIGNIFICANT" if p_val < alpha else "NOT significant"
    print(f"  {source}: F = {anova_table.loc[source, 'F']:.3f}, p = {p_val:.4f} → {sig}")
print(f"\n  The interaction term tells us: the effect of temperature on")
print(f"  strength DEPENDS on the pressure level (non-additive effects).")
print("="*70)

# Interaction Plot
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Interaction plot
for pressure in pressures:
    subset = df_2way[df_2way['Pressure'] == pressure]
    means = subset.groupby('Temperature')['Strength'].mean()
    means = means.reindex(temperatures)
    axes[0].plot(temperatures, means.values, 'o-', linewidth=2, markersize=8, label=f'Pressure: {pressure}')

axes[0].set_xlabel('Temperature')
axes[0].set_ylabel('Mean Tensile Strength (MPa)')
axes[0].set_title('Interaction Plot: Temperature × Pressure')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Grouped bar chart
cell_means_pivot = df_2way.pivot_table(values='Strength', index='Temperature', 
                                        columns='Pressure', aggfunc='mean')
cell_means_pivot = cell_means_pivot.reindex(temperatures)
cell_means_pivot.plot(kind='bar', ax=axes[1], edgecolor='black', alpha=0.8)
axes[1].set_xlabel('Temperature')
axes[1].set_ylabel('Mean Tensile Strength (MPa)')
axes[1].set_title('Cell Means by Factor Combination')
axes[1].legend(title='Pressure')
axes[1].set_xticklabels(temperatures, rotation=0)

plt.tight_layout()
plt.show()
````

## 10. Pearson Correlation Coefficient

### Purpose
Measure the **strength and direction of the linear relationship** between two continuous variables.

### Population Parameter

$$\rho = \frac{\text{Cov}(X, Y)}{\sigma_X \sigma_Y} = \frac{E[(X - \mu_X)(Y - \mu_Y)]}{\sigma_X \sigma_Y}$$

### Sample Statistic

$$r = \frac{\sum_{i=1}^{n}(X_i - \bar{X})(Y_i - \bar{Y})}{\sqrt{\sum_{i=1}^{n}(X_i - \bar{X})^2 \cdot \sum_{i=1}^{n}(Y_i - \bar{Y})^2}}$$

Equivalently:

$$r = \frac{n\sum X_iY_i - \sum X_i \sum Y_i}{\sqrt{[n\sum X_i^2 - (\sum X_i)^2][n\sum Y_i^2 - (\sum Y_i)^2]}}$$

### Properties
- $$-1 \leq r \leq 1$$
- $$r = 1$$: Perfect positive linear relationship
- $$r = -1$$: Perfect negative linear relationship
- $$r = 0$$: No linear relationship (NOT no relationship!)

### Hypothesis Test for $$\rho$$

$$H_0: \rho = 0 \quad (\text{no linear correlation})$$
$$H_1: \rho \neq 0$$

### Test Statistic

$$t = \frac{r\sqrt{n-2}}{\sqrt{1 - r^2}} \sim t_{n-2}$$

### Coefficient of Determination

$$r^2$$ = proportion of variance in $$Y$$ explained by the linear relationship with $$X$$.

### Interpretation Guidelines

| $$|r|$$ | Strength |
| --- | --- |
| 0.00 – 0.19 | Very weak |
| 0.20 – 0.39 | Weak |
| 0.40 – 0.59 | Moderate |
| 0.60 – 0.79 | Strong |
| 0.80 – 1.00 | Very strong |

### Fisher's Z-Transformation (for CI and comparing correlations)

$$z = \frac{1}{2}\ln\left(\frac{1+r}{1-r}\right) = \text{arctanh}(r)$$

$$SE_z = \frac{1}{\sqrt{n-3}}$$

### Assumptions
1. Both variables are continuous (interval/ratio)
2. Linear relationship between variables
3. Bivariate normality (for inference)
4. No extreme outliers
5. Observations are independent pairs

### Industrial Applications
- **Finance**: Correlation between stock returns for portfolio diversification
- **Marketing**: Relationship between ad spend and revenue
- **Healthcare**: Correlation between BMI and blood pressure
- **Engineering**: Relationship between load and deflection in materials
- **Psychology**: Correlation between study hours and exam performance

### Caution: Correlation ≠ Causation!
$$r$$ measures association, not causality. A strong $$r$$ may reflect:
- True causal link (A → B)
- Reverse causation (B → A)
- Confounding variable (C → A and C → B)
- Coincidence (spurious correlation)

````python
# =============================================================================
# PEARSON CORRELATION: Marketing Spend vs Revenue
# =============================================================================
# Scenario: Analyzing the relationship between monthly advertising
# spend ($K) and monthly revenue ($K) for a retail company.

np.random.seed(42)
n_obs = 50

# Generate correlated data
ad_spend = np.random.uniform(10, 100, size=n_obs)
noise = np.random.normal(0, 30, size=n_obs)
revenue = 50 + 2.5 * ad_spend + noise  # True linear relationship + noise

# Calculate Pearson correlation manually
x_dev = ad_spend - ad_spend.mean()
y_dev = revenue - revenue.mean()
r_manual = np.sum(x_dev * y_dev) / np.sqrt(np.sum(x_dev**2) * np.sum(y_dev**2))

# Using scipy
r_scipy, p_scipy = stats.pearsonr(ad_spend, revenue)

# Test statistic
t_corr = r_scipy * np.sqrt(n_obs - 2) / np.sqrt(1 - r_scipy**2)

print("="*60)
print("PEARSON CORRELATION ANALYSIS")
print("="*60)
print(f"\nVariables: Ad Spend ($K) vs Revenue ($K)")
print(f"Sample size: {n_obs}")
print(f"\nPearson r (manual):   {r_manual:.6f}")
print(f"Pearson r (scipy):    {r_scipy:.6f}")
print(f"p-value:              {p_scipy:.8f}")
print(f"t-statistic:          {t_corr:.4f}")
print(f"df:                   {n_obs - 2}")
print(f"\nCoefficient of Determination (r²): {r_scipy**2:.4f}")
print(f"  → {r_scipy**2*100:.1f}% of variance in Revenue is explained")
print(f"    by the linear relationship with Ad Spend.")

# Strength interpretation
abs_r = abs(r_scipy)
if abs_r < 0.2:
    strength = "Very Weak"
elif abs_r < 0.4:
    strength = "Weak"
elif abs_r < 0.6:
    strength = "Moderate"
elif abs_r < 0.8:
    strength = "Strong"
else:
    strength = "Very Strong"

direction = "Positive" if r_scipy > 0 else "Negative"
print(f"\nInterpretation: {strength} {direction} linear relationship")

# Fisher's Z-transformation for CI
z_fisher = np.arctanh(r_scipy)
se_z = 1 / np.sqrt(n_obs - 3)
z_crit_95 = 1.96
ci_z_low = z_fisher - z_crit_95 * se_z
ci_z_high = z_fisher + z_crit_95 * se_z
ci_r_low = np.tanh(ci_z_low)
ci_r_high = np.tanh(ci_z_high)
print(f"\n95% CI for ρ (Fisher's Z): ({ci_r_low:.4f}, {ci_r_high:.4f})")

alpha = 0.05
print(f"\n{'='*60}")
print("INTERPRETATION:")
if p_scipy < alpha:
    print(f"  Since p-value ({p_scipy:.2e}) < α ({alpha}),")
    print(f"  we REJECT H₀: ρ = 0.")
    print(f"  There IS a significant linear correlation between")
    print(f"  ad spend and revenue (r = {r_scipy:.3f}).")
print("="*60)

# Visualization
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# Scatter plot with regression line
axes[0].scatter(ad_spend, revenue, alpha=0.6, edgecolors='black', linewidth=0.5)
z = np.polyfit(ad_spend, revenue, 1)
p_line = np.poly1d(z)
x_sorted = np.sort(ad_spend)
axes[0].plot(x_sorted, p_line(x_sorted), 'r-', linewidth=2, 
             label=f'r = {r_scipy:.3f}, r² = {r_scipy**2:.3f}')
axes[0].set_xlabel('Ad Spend ($K)')
axes[0].set_ylabel('Revenue ($K)')
axes[0].set_title('Scatter Plot with Regression Line')
axes[0].legend(fontsize=11)

# Residual plot
residuals = revenue - p_line(ad_spend)
axes[1].scatter(ad_spend, residuals, alpha=0.6, edgecolors='black', linewidth=0.5)
axes[1].axhline(0, color='red', linestyle='--')
axes[1].set_xlabel('Ad Spend ($K)')
axes[1].set_ylabel('Residuals')
axes[1].set_title('Residual Plot (Check for Patterns)')

# Correlation heatmap example with multiple variables
np.random.seed(42)
df_corr = pd.DataFrame({
    'Ad_Spend': ad_spend,
    'Revenue': revenue,
    'Web_Traffic': 1000 + 15*ad_spend + np.random.normal(0, 100, n_obs),
    'Customer_Satisfaction': 3 + 0.01*revenue + np.random.normal(0, 0.5, n_obs)
})
corr_matrix = df_corr.corr()
sns.heatmap(corr_matrix, annot=True, cmap='RdBu_r', center=0, vmin=-1, vmax=1,
            ax=axes[2], fmt='.3f', square=True)
axes[2].set_title('Correlation Heatmap')

plt.tight_layout()
plt.show()
````

## 11. Simple Linear Regression

### Purpose
Model the **linear relationship** between a continuous predictor ($$X$$) and a continuous response ($$Y$$), and test whether the relationship is statistically significant.

### The Model

$$Y_i = \beta_0 + \beta_1 X_i + \epsilon_i$$

Where:
- $$\beta_0$$ = intercept (expected value of $$Y$$ when $$X = 0$$)
- $$\beta_1$$ = slope (change in $$Y$$ per unit change in $$X$$)
- $$\epsilon_i \sim N(0, \sigma^2)$$ = random error (independently and identically distributed)

### Ordinary Least Squares (OLS) Estimation

Minimize the sum of squared residuals:

$$\min_{\beta_0, \beta_1} \sum_{i=1}^{n} (Y_i - \beta_0 - \beta_1 X_i)^2$$

Solutions:

$$\hat{\beta}_1 = \frac{\sum_{i=1}^{n}(X_i - \bar{X})(Y_i - \bar{Y})}{\sum_{i=1}^{n}(X_i - \bar{X})^2} = \frac{S_{XY}}{S_{XX}}$$

$$\hat{\beta}_0 = \bar{Y} - \hat{\beta}_1 \bar{X}$$

### Hypothesis Tests for Regression Coefficients

**Test for slope (most important):**

$$H_0: \beta_1 = 0 \quad (\text{no linear relationship})$$
$$H_1: \beta_1 \neq 0$$

Test statistic:

$$t = \frac{\hat{\beta}_1}{SE(\hat{\beta}_1)} \sim t_{n-2}$$

Where:

$$SE(\hat{\beta}_1) = \frac{s}{\sqrt{S_{XX}}} = \frac{\sqrt{\frac{\sum(Y_i - \hat{Y}_i)^2}{n-2}}}{\sqrt{\sum(X_i - \bar{X})^2}}$$

### ANOVA for Regression

$$SS_{\text{Total}} = SS_{\text{Regression}} + SS_{\text{Residual}}$$

$$R^2 = \frac{SS_{\text{Reg}}}{SS_{\text{Total}}} = 1 - \frac{SS_{\text{Res}}}{SS_{\text{Total}}}$$

The overall F-test:

$$F = \frac{MS_{\text{Reg}}}{MS_{\text{Res}}} = \frac{SS_{\text{Reg}} / 1}{SS_{\text{Res}} / (n-2)} \sim F_{1, n-2}$$

**Note**: For simple linear regression, $$F = t^2$$ (the F-test and t-test for $$\beta_1$$ are equivalent).

### Assumptions (LINE)
1. **L**inearity: $$E[Y|X]$$ is a linear function of $$X$$
2. **I**ndependence: Errors are independent
3. **N**ormality: Errors are normally distributed ($$\epsilon_i \sim N(0, \sigma^2)$$)
4. **E**qual variance (homoscedasticity): $$\text{Var}(\epsilon_i) = \sigma^2$$ for all $$i$$

### Diagnostic Checks
- **Residuals vs. Fitted**: Check linearity and homoscedasticity
- **Q-Q Plot of Residuals**: Check normality
- **Scale-Location Plot**: Check homoscedasticity
- **Residuals vs. Leverage**: Identify influential points

### Industrial Applications
- **Real Estate**: Predicting house price from square footage
- **Engineering**: Modeling stress-strain relationships
- **Economics**: Relating GDP growth to unemployment (Okun's Law)
- **Environmental Science**: Modeling pollutant concentration vs. distance from source
- **Marketing**: Modeling sales as a function of price

````python
# =============================================================================
# SIMPLE LINEAR REGRESSION: Real Estate Price Prediction
# =============================================================================
# Scenario: Modeling house prices based on square footage.

np.random.seed(42)
n_houses = 80

# Generate data
sqft = np.random.uniform(800, 3500, size=n_houses)
price = 50000 + 150 * sqft + np.random.normal(0, 40000, size=n_houses)  # $150 per sqft + noise

# Fit regression using statsmodels
X_sm = sm.add_constant(sqft)  # Add intercept
model_reg = sm.OLS(price, X_sm).fit()

print("="*70)
print("SIMPLE LINEAR REGRESSION: House Price ~ Square Footage")
print("="*70)
print(model_reg.summary())

# Extract key results
print(f"\n{'='*70}")
print("KEY RESULTS SUMMARY")
print(f"{'='*70}")
print(f"\nRegression Equation: Price = {model_reg.params[0]:.2f} + {model_reg.params[1]:.2f} × SqFt")
print(f"\nIntercept (β₀):     {model_reg.params[0]:>12.2f} (p = {model_reg.pvalues[0]:.4f})")
print(f"Slope (β₁):         {model_reg.params[1]:>12.2f} (p = {model_reg.pvalues[1]:.2e})")
print(f"\nR²:                 {model_reg.rsquared:.4f} ({model_reg.rsquared*100:.1f}% variance explained)")
print(f"Adjusted R²:        {model_reg.rsquared_adj:.4f}")
print(f"F-statistic:        {model_reg.fvalue:.4f} (p = {model_reg.f_pvalue:.2e})")
print(f"Residual Std Error: {np.sqrt(model_reg.mse_resid):.2f}")

print(f"\n{'='*70}")
print("INTERPRETATION:")
print(f"  For every additional square foot, the house price increases")
print(f"  by approximately ${model_reg.params[1]:.2f} on average.")
print(f"  The model explains {model_reg.rsquared*100:.1f}% of the variation in house prices.")
if model_reg.pvalues[1] < 0.05:
    print(f"  The relationship is statistically significant (p < 0.001).")
print("="*70)

# Diagnostic Plots
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Scatter + Regression line
axes[0, 0].scatter(sqft, price, alpha=0.5, edgecolors='black', linewidth=0.5)
axes[0, 0].plot(np.sort(sqft), model_reg.predict(sm.add_constant(np.sort(sqft))), 
               'r-', linewidth=2, label=f'y = {model_reg.params[0]:.0f} + {model_reg.params[1]:.0f}x')
axes[0, 0].fill_between(np.sort(sqft), 
                         model_reg.predict(sm.add_constant(np.sort(sqft))) - 2*np.sqrt(model_reg.mse_resid),
                         model_reg.predict(sm.add_constant(np.sort(sqft))) + 2*np.sqrt(model_reg.mse_resid),
                         alpha=0.1, color='red', label='±2σ band')
axes[0, 0].set_xlabel('Square Footage')
axes[0, 0].set_ylabel('Price ($)')
axes[0, 0].set_title(f'Linear Regression (R² = {model_reg.rsquared:.3f})')
axes[0, 0].legend()

# 2. Residuals vs Fitted
fitted = model_reg.fittedvalues
residuals_reg = model_reg.resid
axes[0, 1].scatter(fitted, residuals_reg, alpha=0.5, edgecolors='black', linewidth=0.5)
axes[0, 1].axhline(0, color='red', linestyle='--')
axes[0, 1].set_xlabel('Fitted Values')
axes[0, 1].set_ylabel('Residuals')
axes[0, 1].set_title('Residuals vs Fitted (Check Linearity & Homoscedasticity)')

# 3. Q-Q Plot
stats.probplot(residuals_reg, dist="norm", plot=axes[1, 0])
axes[1, 0].set_title('Q-Q Plot of Residuals (Check Normality)')

# 4. Scale-Location
std_resid = residuals_reg / np.sqrt(model_reg.mse_resid)
axes[1, 1].scatter(fitted, np.sqrt(np.abs(std_resid)), alpha=0.5, edgecolors='black', linewidth=0.5)
axes[1, 1].set_xlabel('Fitted Values')
axes[1, 1].set_ylabel('√|Standardized Residuals|')
axes[1, 1].set_title('Scale-Location Plot (Check Homoscedasticity)')

plt.tight_layout()
plt.show()
````

## 12. Chi-Square Test for Variance

### Purpose
Test whether the population variance $$\sigma^2$$ equals a hypothesized value $$\sigma_0^2$$.

### Hypotheses

$$H_0: \sigma^2 = \sigma_0^2$$
$$H_1: \sigma^2 \neq \sigma_0^2 \quad (\text{or } > \text{ or } <)$$

### Test Statistic

$$\chi^2 = \frac{(n-1)s^2}{\sigma_0^2} \sim \chi^2_{n-1}$$

### The Chi-Square Distribution

If $$Z_1, Z_2, \ldots, Z_k$$ are independent standard normal variables, then:

$$\chi^2_k = \sum_{i=1}^{k} Z_i^2$$

Properties:
- Always non-negative (sum of squares)
- Right-skewed (becomes more symmetric as $$df$$ increases)
- Mean = $$k$$, Variance = $$2k$$

### Confidence Interval for $$\sigma^2$$

$$\left(\frac{(n-1)s^2}{\chi^2_{\alpha/2, n-1}}, \quad \frac{(n-1)s^2}{\chi^2_{1-\alpha/2, n-1}}\right)$$

### Assumptions
1. Data comes from a **normally distributed** population (this test is very sensitive to non-normality!)
2. Random sampling
3. Continuous data

### Industrial Applications
- **Manufacturing**: Testing if machine output variability meets precision specifications
- **Quality Control**: Verifying that batch-to-batch variation in pharmaceutical concentration is within limits
- **Finance**: Testing if portfolio volatility (variance of returns) matches a target risk level
- **Metrology**: Checking if measurement instrument precision meets calibration standards

## 13. F-Test for Equality of Variances

### Purpose
Compare the variances of **two independent populations** to test if they are equal.

### Hypotheses

$$H_0: \sigma_1^2 = \sigma_2^2 \quad (\text{or equivalently } \frac{\sigma_1^2}{\sigma_2^2} = 1)$$
$$H_1: \sigma_1^2 \neq \sigma_2^2$$

### Test Statistic

$$F = \frac{s_1^2}{s_2^2} \sim F_{n_1-1, \; n_2-1}$$

Convention: Place the larger variance in the numerator so $$F \geq 1$$.

### The F-Distribution

If $$U \sim \chi^2_{d_1}$$ and $$V \sim \chi^2_{d_2}$$ are independent, then:

$$F = \frac{U/d_1}{V/d_2} \sim F_{d_1, d_2}$$

Properties:
- Always non-negative
- Right-skewed
- Mean $$\approx 1$$ when $$H_0$$ is true (exactly $$\frac{d_2}{d_2-2}$$ for $$d_2 > 2$$)

### Relationship to Other Tests
- ANOVA's F-statistic uses the same F-distribution
- For two groups: $$F_{\text{ANOVA}} = t^2$$ (t-test squared)

### Assumptions
1. Both populations are **normally distributed** (very sensitive!)
2. Samples are independent
3. Random sampling

### Industrial Applications
- **Pre-test for t-test**: Deciding between Student's and Welch's t-test
- **Manufacturing**: Comparing consistency of two production processes
- **Finance**: Comparing volatility between two investment strategies
- **Laboratory**: Comparing precision of two measurement methods

````python
# =============================================================================
# CHI-SQUARE TEST FOR VARIANCE & F-TEST FOR EQUALITY OF VARIANCES
# =============================================================================

# --- Part 1: Chi-Square Test for Variance ---
# Scenario: A machine should produce bolts with diameter variance ≤ 0.04 mm².
# We sample 25 bolts and test if the variance exceeds the specification.

print("="*70)
print("PART 1: CHI-SQUARE TEST FOR VARIANCE")
print("="*70)

np.random.seed(42)
n_bolts = 25
sigma_0_sq = 0.04  # Hypothesized variance (specification)
bolts = np.random.normal(loc=10.0, scale=0.25, size=n_bolts)  # True σ = 0.25, σ² = 0.0625

s_sq = bolts.var(ddof=1)  # Sample variance
chi2_stat = (n_bolts - 1) * s_sq / sigma_0_sq
df_chi = n_bolts - 1

# Right-tailed test (testing if variance EXCEEDS spec)
p_right = 1 - stats.chi2.cdf(chi2_stat, df=df_chi)
# Two-tailed test
p_two = 2 * min(stats.chi2.cdf(chi2_stat, df=df_chi), 1 - stats.chi2.cdf(chi2_stat, df=df_chi))

print(f"\nSample size:              {n_bolts}")
print(f"Sample variance (s²):     {s_sq:.6f} mm²")
print(f"Hypothesized σ₀²:          {sigma_0_sq} mm²")
print(f"\nχ² statistic:             {chi2_stat:.4f}")
print(f"Degrees of freedom:       {df_chi}")
print(f"p-value (right-tailed):   {p_right:.6f}")
print(f"p-value (two-tailed):     {p_two:.6f}")

# CI for population variance
alpha = 0.05
chi2_lower = stats.chi2.ppf(alpha/2, df=df_chi)
chi2_upper = stats.chi2.ppf(1 - alpha/2, df=df_chi)
ci_var_lower = (df_chi * s_sq) / chi2_upper
ci_var_upper = (df_chi * s_sq) / chi2_lower
print(f"\n95% CI for σ²:            ({ci_var_lower:.6f}, {ci_var_upper:.6f})")
print(f"95% CI for σ:             ({np.sqrt(ci_var_lower):.6f}, {np.sqrt(ci_var_upper):.6f})")

print(f"\nINTERPRETATION:")
if p_right < alpha:
    print(f"  REJECT H₀: The process variance significantly exceeds the")
    print(f"  specification of {sigma_0_sq} mm². Machine needs recalibration.")
else:
    print(f"  FAIL TO REJECT H₀: No evidence variance exceeds spec.")

# --- Part 2: F-Test for Equality of Variances ---
print(f"\n\n{'='*70}")
print("PART 2: F-TEST FOR EQUALITY OF VARIANCES")
print("="*70)

# Scenario: Compare precision of two manufacturing machines
np.random.seed(42)
machine_1 = np.random.normal(loc=50, scale=3.0, size=30)  # σ² = 9
machine_2 = np.random.normal(loc=50, scale=4.5, size=35)  # σ² = 20.25

s1_sq = machine_1.var(ddof=1)
s2_sq = machine_2.var(ddof=1)

# Place larger variance in numerator
if s1_sq >= s2_sq:
    f_stat = s1_sq / s2_sq
    df1_f, df2_f = len(machine_1) - 1, len(machine_2) - 1
else:
    f_stat = s2_sq / s1_sq
    df1_f, df2_f = len(machine_2) - 1, len(machine_1) - 1

p_f_two = 2 * (1 - stats.f.cdf(f_stat, df1_f, df2_f))

print(f"\nMachine 1: n={len(machine_1)}, s²={s1_sq:.4f}")
print(f"Machine 2: n={len(machine_2)}, s²={s2_sq:.4f}")
print(f"\nF-statistic:    {f_stat:.4f}")
print(f"df:             ({df1_f}, {df2_f})")
print(f"p-value:        {p_f_two:.6f}")
print(f"F-critical (α=0.05): {stats.f.ppf(0.975, df1_f, df2_f):.4f}")

# Also compare with Levene's (more robust)
lev_f, lev_p = stats.levene(machine_1, machine_2)
print(f"\nLevene's Test (robust alternative):")
print(f"  F = {lev_f:.4f}, p = {lev_p:.6f}")

print(f"\nINTERPRETATION:")
if p_f_two < alpha:
    print(f"  REJECT H₀: The two machines have significantly different")
    print(f"  precision (variances). Machine 2 is less consistent.")
    print(f"  Use Welch's t-test (not Student's) for comparing their means.")
else:
    print(f"  FAIL TO REJECT H₀: No evidence of different variances.")
print("="*70)

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Chi-square distribution
x_chi = np.linspace(0, 60, 1000)
axes[0].plot(x_chi, stats.chi2.pdf(x_chi, df=df_chi), 'b-', linewidth=2)
axes[0].axvline(chi2_stat, color='red', linestyle='--', linewidth=2, label=f'χ² = {chi2_stat:.2f}')
crit_right = stats.chi2.ppf(0.95, df=df_chi)
axes[0].fill_between(x_chi[x_chi >= crit_right], stats.chi2.pdf(x_chi[x_chi >= crit_right], df=df_chi),
                     alpha=0.3, color='red', label=f'Rejection (χ² > {crit_right:.2f})')
axes[0].set_xlabel('χ² value')
axes[0].set_ylabel('Density')
axes[0].set_title(f'Chi-Square Distribution (df={df_chi})')
axes[0].legend()

# F-distribution
x_f = np.linspace(0, 5, 1000)
axes[1].plot(x_f, stats.f.pdf(x_f, df1_f, df2_f), 'b-', linewidth=2)
axes[1].axvline(f_stat, color='red', linestyle='--', linewidth=2, label=f'F = {f_stat:.2f}')
f_crit = stats.f.ppf(0.975, df1_f, df2_f)
axes[1].fill_between(x_f[x_f >= f_crit], stats.f.pdf(x_f[x_f >= f_crit], df1_f, df2_f),
                     alpha=0.3, color='red', label=f'Rejection (F > {f_crit:.2f})')
axes[1].set_xlabel('F value')
axes[1].set_ylabel('Density')
axes[1].set_title(f'F-Distribution (df1={df1_f}, df2={df2_f})')
axes[1].legend()

plt.tight_layout()
plt.show()
````

## 14. Statistical Power & Sample Size Determination

### The Four Pillars of Hypothesis Testing

Every hypothesis test involves a trade-off between four quantities:

| Quantity | Symbol | Description |
| --- | --- | --- |
| Significance Level | $$\alpha$$ | P(Type I Error) = P(Reject $$H_0$$ \| $$H_0$$ true) |
| Power | $$1 - \beta$$ | P(Reject $$H_0$$ \| $$H_1$$ true) |
| Effect Size | $$d, f, r$$ | Magnitude of the real difference |
| Sample Size | $$n$$ | Number of observations |

**Fix any three, and the fourth is determined.**

### Type I and Type II Errors

| | $$H_0$$ True | $$H_0$$ False |
| --- | --- | --- |
| **Fail to Reject $$H_0$$** | Correct ($$1-\alpha$$) | Type II Error ($$\beta$$) |
| **Reject $$H_0$$** | Type I Error ($$\alpha$$) | Correct = Power ($$1-\beta$$) |

### Power Formulas

For a two-sample t-test (equal $$n$$):

$$n = \frac{2(Z_{\alpha/2} + Z_{\beta})^2 \sigma^2}{\delta^2}$$

Where $$\delta = \mu_1 - \mu_2$$ is the minimum detectable difference.

Equivalently using Cohen's $$d$$:

$$n = \frac{2(Z_{\alpha/2} + Z_{\beta})^2}{d^2}$$

### Why Power Analysis Matters

- **Before a study**: Determine the minimum $$n$$ needed to detect a meaningful effect
- **After a study**: If $$H_0$$ is not rejected, was the study **underpowered**? (Low power → high $$\beta$$ → might miss a real effect)
- **Industry standard**: Power $$\geq 0.80$$ (80% chance of detecting a true effect)

````python
# =============================================================================
# STATISTICAL POWER ANALYSIS & SAMPLE SIZE DETERMINATION
# =============================================================================
from statsmodels.stats.power import TTestIndPower, TTestPower, FTestAnovaPower

print("="*70)
print("POWER ANALYSIS FOR PARAMETRIC TESTS")
print("="*70)

# --- Example 1: Two-sample t-test sample size ---
print("\n--- Required Sample Size: Two-Sample t-Test ---")
power_analysis = TTestIndPower()

# Parameters
effect_sizes = [0.2, 0.5, 0.8]  # Small, Medium, Large
alpha_pa = 0.05
power_target = 0.80

print(f"\n{'Effect Size (d)':<18} {'Required n (per group)':<25} {'Total N'}")
print("-"*55)
for d in effect_sizes:
    n_required = power_analysis.solve_power(effect_size=d, alpha=alpha_pa, 
                                             power=power_target, alternative='two-sided')
    print(f"{d:<18} {n_required:<25.0f} {2*n_required:.0f}")

# --- Example 2: Power curves ---
print("\n--- Power as a Function of Sample Size ---")
print(f"(Effect size d=0.5, α=0.05, two-sided)")
print(f"\n{'n per group':<15} {'Power'}")
print("-"*30)
for n_test in [10, 20, 30, 50, 80, 100, 150, 200]:
    power_val = power_analysis.solve_power(effect_size=0.5, nobs1=n_test, 
                                            alpha=0.05, alternative='two-sided')
    bar = '█' * int(power_val * 40)
    print(f"{n_test:<15} {power_val:.4f}  {bar}")

# --- Example 3: One-sample t-test ---
print("\n\n--- Required Sample Size: One-Sample t-Test ---")
power_1samp = TTestPower()
for d in effect_sizes:
    n_req = power_1samp.solve_power(effect_size=d, alpha=0.05, power=0.80, alternative='two-sided')
    print(f"  d = {d}: n = {n_req:.0f}")

# Visualization: Power Curves
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Power vs Sample Size for different effect sizes
n_range = np.arange(5, 200)
for d in [0.2, 0.5, 0.8]:
    powers = [power_analysis.solve_power(effect_size=d, nobs1=n, alpha=0.05, alternative='two-sided') 
              for n in n_range]
    axes[0].plot(n_range, powers, linewidth=2, label=f'd = {d} ({["Small","Medium","Large"][[0.2,0.5,0.8].index(d)]})')

axes[0].axhline(0.80, color='red', linestyle='--', alpha=0.7, label='Target Power = 0.80')
axes[0].set_xlabel('Sample Size (per group)')
axes[0].set_ylabel('Statistical Power')
axes[0].set_title('Power Curves: Two-Sample t-Test (α = 0.05)')
axes[0].legend()
axes[0].set_ylim(0, 1)
axes[0].grid(True, alpha=0.3)

# Power vs Effect Size for different sample sizes
d_range = np.linspace(0.1, 1.5, 100)
for n_val in [20, 50, 100, 200]:
    powers = [power_analysis.solve_power(effect_size=d, nobs1=n_val, alpha=0.05, alternative='two-sided') 
              for d in d_range]
    axes[1].plot(d_range, powers, linewidth=2, label=f'n = {n_val}')

axes[1].axhline(0.80, color='red', linestyle='--', alpha=0.7, label='Target Power = 0.80')
axes[1].set_xlabel("Cohen's d (Effect Size)")
axes[1].set_ylabel('Statistical Power')
axes[1].set_title('Power vs Effect Size for Different Sample Sizes')
axes[1].legend()
axes[1].set_ylim(0, 1)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

````python
# =============================================================================
# COMPREHENSIVE ASSUMPTION CHECKING TOOLKIT
# =============================================================================
# A reusable function to check all parametric test assumptions

def check_parametric_assumptions(data, group_labels=None, alpha=0.05):
    """
    Comprehensive assumption checker for parametric tests.
    
    Parameters:
    -----------
    data : array-like or list of arrays
        Single array for one-sample tests, list of arrays for multi-group
    group_labels : list of str, optional
        Names for each group
    alpha : float
        Significance level for assumption tests
    """
    print("="*70)
    print("PARAMETRIC ASSUMPTIONS DIAGNOSTIC REPORT")
    print("="*70)
    
    if not isinstance(data, list):
        data = [data]
    
    if group_labels is None:
        group_labels = [f'Group {i+1}' for i in range(len(data))]
    
    k = len(data)
    
    # 1. NORMALITY CHECK
    print("\n" + "-"*40)
    print("1. NORMALITY (Shapiro-Wilk Test)")
    print("-"*40)
    all_normal = True
    for i, (d, label) in enumerate(zip(data, group_labels)):
        if len(d) >= 3:  # Shapiro-Wilk needs n >= 3
            stat, p = stats.shapiro(d)
            normal = p > alpha
            all_normal = all_normal and normal
            symbol = '\u2713' if normal else '\u2717'
            print(f"  {label:<20}: W = {stat:.4f}, p = {p:.4f} {symbol} {'Normal' if normal else 'NON-NORMAL'}")
            # Skewness and Kurtosis
            skew = stats.skew(d)
            kurt = stats.kurtosis(d)  # Excess kurtosis
            print(f"  {'':20}  Skewness = {skew:.3f}, Excess Kurtosis = {kurt:.3f}")
    
    # 2. HOMOGENEITY OF VARIANCE (for multiple groups)
    if k > 1:
        print("\n" + "-"*40)
        print("2. HOMOGENEITY OF VARIANCE")
        print("-"*40)
        
        # Levene's test
        lev_stat, lev_p = stats.levene(*data)
        equal_var = lev_p > alpha
        print(f"  Levene's Test: F = {lev_stat:.4f}, p = {lev_p:.4f} {'\u2713' if equal_var else '\u2717'} {'Equal' if equal_var else 'UNEQUAL'} variances")
        
        # Bartlett's test (assumes normality)
        if all_normal:
            bart_stat, bart_p = stats.bartlett(*data)
            print(f"  Bartlett's Test: χ² = {bart_stat:.4f}, p = {bart_p:.4f} {'\u2713' if bart_p > alpha else '\u2717'}")
        
        # Variance ratio
        variances = [d.var(ddof=1) for d in data]
        print(f"  Variance ratio (max/min): {max(variances)/min(variances):.2f}")
        print(f"  Rule of thumb: ratio < 4 is acceptable for ANOVA")
    
    # 3. OUTLIER CHECK
    print("\n" + "-"*40)
    print("3. OUTLIER CHECK (IQR Method)")
    print("-"*40)
    for d, label in zip(data, group_labels):
        q1, q3 = np.percentile(d, [25, 75])
        iqr = q3 - q1
        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr
        outliers = d[(d < lower) | (d > upper)]
        print(f"  {label:<20}: {len(outliers)} outliers detected (of {len(d)} observations)")
    
    # 4. SAMPLE SIZE
    print("\n" + "-"*40)
    print("4. SAMPLE SIZE ADEQUACY")
    print("-"*40)
    for d, label in zip(data, group_labels):
        n = len(d)
        adequate = n >= 30
        print(f"  {label:<20}: n = {n} {'\u2713 CLT applies' if adequate else '\u26a0 Small sample - normality critical'}")
    
    # OVERALL RECOMMENDATION
    print("\n" + "="*70)
    print("RECOMMENDATION:")
    if all_normal and (k == 1 or equal_var):
        print("  \u2713 All assumptions met. Parametric test is appropriate.")
    elif not all_normal:
        print("  \u26a0 Normality violated. Consider:")
        print("    - Non-parametric alternative (Mann-Whitney, Kruskal-Wallis)")
        print("    - Data transformation (log, sqrt, Box-Cox)")
        print("    - Bootstrapping")
        if all(len(d) >= 30 for d in data):
            print("    - However, n ≥ 30 in all groups (CLT), so parametric may still be OK.")
    elif k > 1 and not equal_var:
        print("  \u26a0 Unequal variances. Use:")
        print("    - Welch's t-test (for 2 groups)")
        print("    - Welch's ANOVA (for 3+ groups)")
        print("    - Games-Howell post-hoc test")
    print("="*70)

# Demo: Check assumptions for our ANOVA data
print("\nDEMO: Checking assumptions for the Teaching Methods data\n")
check_parametric_assumptions(
    [method_a, method_b, method_c, method_d],
    ['Traditional', 'Flipped', 'Online', 'Hybrid']
)
````

## 15. Summary: Choosing the Right Parametric Test

### Decision Flowchart

```
                    How many groups/conditions?
                            |
            ┌───────────────┼───────────────┐
            1 sample         2 samples         3+ samples
            |                |                  |
      σ known?         Independent?         How many factors?
       /     \          /        \            /          \
     Yes     No       Yes        No        1 factor    2+ factors
      |       |        |          |          |            |
   Z-test  t-test  Ind. t-test  Paired   One-Way      Two-Way
                                t-test    ANOVA        ANOVA
```

### Quick Reference Table

| Test | Purpose | Test Statistic | Distribution | Key Assumption |
| --- | --- | --- | --- | --- |
| **Z-test** | 1 mean, $$\sigma$$ known | $$Z = \frac{\bar{X}-\mu_0}{\sigma/\sqrt{n}}$$ | $$N(0,1)$$ | Known $$\sigma$$ |
| **One-sample t** | 1 mean, $$\sigma$$ unknown | $$t = \frac{\bar{X}-\mu_0}{s/\sqrt{n}}$$ | $$t_{n-1}$$ | Normality |
| **Independent t** | 2 independent means | $$t = \frac{\bar{X}_1-\bar{X}_2}{SE}$$ | $$t_{df}$$ | Independence, equal variance |
| **Paired t** | 2 related means | $$t = \frac{\bar{D}}{s_D/\sqrt{n}}$$ | $$t_{n-1}$$ | Normal differences |
| **One-Way ANOVA** | 3+ means, 1 factor | $$F = \frac{MS_B}{MS_W}$$ | $$F_{k-1, N-k}$$ | Homoscedasticity |
| **Two-Way ANOVA** | Means, 2 factors | Multiple F-ratios | $$F_{df_1, df_2}$$ | Balanced design |
| **Pearson r** | Linear correlation | $$t = \frac{r\sqrt{n-2}}{\sqrt{1-r^2}}$$ | $$t_{n-2}$$ | Bivariate normality |
| **Linear Regression** | Predict Y from X | $$t = \frac{\hat{\beta}}{SE(\hat{\beta})}$$ | $$t_{n-2}$$ | LINE assumptions |
| **Chi-square (variance)** | 1 variance | $$\chi^2 = \frac{(n-1)s^2}{\sigma_0^2}$$ | $$\chi^2_{n-1}$$ | Strict normality |
| **F-test (variances)** | 2 variances | $$F = \frac{s_1^2}{s_2^2}$$ | $$F_{n_1-1, n_2-1}$$ | Strict normality |

### When Assumptions Fail: Parametric Alternatives

| Issue | Solution |
| --- | --- |
| Non-normality | Transform data (log, sqrt, Box-Cox), use non-parametric test, or bootstrap |
| Unequal variances | Welch's t-test, Welch's ANOVA, or robust methods |
| Outliers | Trim/Winsorize, use robust estimators, or switch to non-parametric |
| Non-independence | Use repeated measures ANOVA, mixed models, or GEE |
| Non-linearity (regression) | Polynomial regression, GAMs, or non-parametric regression |
| Heteroscedasticity (regression) | Weighted LS, robust standard errors (HC), or transform Y |
