# 1. Historical Evolution of Regression

## 1.1 Etymology & Origins

The term **"regression"** was coined by **Sir Francis Galton** in his 1886 paper *"Regression Towards Mediocrity in Hereditary Stature"*. Galton observed that children of exceptionally tall (or short) parents tended to be closer to the population average height — they "regressed" toward the mean.

## 1.2 Timeline of Key Developments

| Year | Contributor | Contribution |
| --- | --- | --- |
| 1805 | Adrien-Marie Legendre | First published the **method of least squares** for fitting lines to astronomical data |
| 1809 | Carl Friedrich Gauss | Independently developed least squares; connected it to the **normal distribution** of errors |
| 1886 | Francis Galton | Coined "regression"; applied it to heredity studies |
| 1897 | Karl Pearson | Formalized the **correlation coefficient** $$r$$ and extended regression to multiple variables |
| 1922 | R.A. Fisher | Developed the framework for **maximum likelihood estimation** (MLE), ANOVA, and hypothesis testing in regression |
| 1943 | Haavelmo | Introduced **probabilistic foundations** to econometric regression models |
| 1970 | Hoerl & Kennard | Introduced **Ridge Regression** ($$L_2$$ regularization) |
| 1996 | Robert Tibshirani | Introduced **LASSO** ($$L_1$$ regularization) |
| 2005 | Zou & Hastie | Introduced **Elastic Net** (combined $$L_1 + L_2$$) |

## 1.3 The Gauss-Legendre Debate

Both Legendre (1805) and Gauss (1809) independently developed least squares. Gauss claimed he had used the method since 1795 but hadn't published. Regardless of priority, **Gauss's contribution was deeper**: he proved that under normally distributed errors, least squares provides the **maximum likelihood estimate** of the regression parameters — establishing the probabilistic backbone of modern regression.

# 2. The Problem Regression Solves

## 2.1 Core Objective

Regression answers the fundamental question:

> **Given a set of input variables (features), how can we best predict or explain the value of a continuous output variable (target)?**

Formally, given a dataset of $$n$$ observations $$\{(x_i, y_i)\}_{i=1}^{n}$$, regression seeks a function $$f$$ such that:

$$y_i = f(x_i) + \varepsilon_i$$

where $$\varepsilon_i$$ represents irreducible noise.

## 2.2 Two Complementary Goals

### Prediction
Estimate $$\hat{y}$$ for new, unseen inputs $$x_{\text{new}}$$ with minimal error.

### Inference
Understand and quantify the **relationship** between $$x$$ and $$y$$:
- *Which* variables matter?
- *How much* does $$y$$ change when $$x_j$$ changes by one unit?
- *Is the relationship statistically significant?*

## 2.3 Why Not Just Use the Mean?

The simplest "model" predicts $$\hat{y} = \bar{y}$$ for all inputs. Regression improves upon this by **explaining variance** in $$y$$ using information from $$x$$. The improvement is measured by $$R^2$$:

$$R^2 = 1 - \frac{\sum_{i=1}^{n}(y_i - \hat{y}_i)^2}{\sum_{i=1}^{n}(y_i - \bar{y})^2} = 1 - \frac{\text{SS}_{\text{res}}}{\text{SS}_{\text{tot}}}$$

## 2.4 Use Cases Across Domains

| Domain | Example |
| --- | --- |
| Economics | Predicting GDP growth from unemployment, inflation, interest rates |
| Medicine | Modeling drug dosage effect on blood pressure |
| Engineering | Predicting material stress from applied force and temperature |
| Marketing | Estimating ad spend ROI on revenue |
| Climate Science | Modeling temperature trends from CO₂ concentration |

# 3. Mathematical Framework & Notation

## 3.1 Standard Notation

| Symbol | Meaning |
| --- | --- |
| $$n$$ | Number of observations (samples) |
| $$p$$ | Number of predictor variables (features) |
| $$y_i$$ | Observed response for the $$i$$-th sample |
| $$\hat{y}_i$$ | Predicted response for the $$i$$-th sample |
| $$x_{ij}$$ | Value of the $$j$$-th feature for the $$i$$-th sample |
| $$\beta_0$$ | Intercept (bias term) |
| $$\beta_j$$ | Coefficient for the $$j$$-th predictor |
| $$\varepsilon_i$$ | Error term (residual) for the $$i$$-th observation |
| $$\hat{\beta}$$ | Estimated (fitted) coefficient vector |
| $$\sigma^2$$ | Variance of the error term |

## 3.2 Matrix Formulation

For $$n$$ observations and $$p$$ predictors, we define:

$$\mathbf{y} = \begin{bmatrix} y_1 \\ y_2 \\ \vdots \\ y_n \end{bmatrix}_{n \times 1}, \quad \mathbf{X} = \begin{bmatrix} 1 & x_{11} & x_{12} & \cdots & x_{1p} \\ 1 & x_{21} & x_{22} & \cdots & x_{2p} \\ \vdots & \vdots & \vdots & \ddots & \vdots \\ 1 & x_{n1} & x_{n2} & \cdots & x_{np} \end{bmatrix}_{n \times (p+1)}, \quad \boldsymbol{\beta} = \begin{bmatrix} \beta_0 \\ \beta_1 \\ \vdots \\ \beta_p \end{bmatrix}_{(p+1) \times 1}$$

The model in matrix form:

$$\mathbf{y} = \mathbf{X}\boldsymbol{\beta} + \boldsymbol{\varepsilon}$$

where $$\boldsymbol{\varepsilon} \sim \mathcal{N}(\mathbf{0}, \sigma^2 \mathbf{I}_n)$$

## 3.3 The Geometry of Regression

Regression finds the vector $$\hat{\mathbf{y}} = \mathbf{X}\hat{\boldsymbol{\beta}}$$ that is the **orthogonal projection** of $$\mathbf{y}$$ onto the column space of $$\mathbf{X}$$. The residual vector $$\mathbf{e} = \mathbf{y} - \hat{\mathbf{y}}$$ is perpendicular to every column of $$\mathbf{X}$$:

$$\mathbf{X}^T(\mathbf{y} - \mathbf{X}\hat{\boldsymbol{\beta}}) = \mathbf{0}$$

This geometric insight explains why OLS minimizes the sum of squared residuals — it finds the closest point in the model subspace.

# 4. Types of Regression

---

## 4.1 Simple Linear Regression (SLR)

### Definition

SLR models the relationship between **one** independent variable $$x$$ and a dependent variable $$y$$ using a straight line.

### The Model

$$y_i = \beta_0 + \beta_1 x_i + \varepsilon_i, \quad i = 1, 2, \ldots, n$$

where:
- $$\beta_0$$ = y-intercept (value of $$y$$ when $$x = 0$$)
- $$\beta_1$$ = slope (change in $$y$$ per unit change in $$x$$)
- $$\varepsilon_i \sim \mathcal{N}(0, \sigma^2)$$ = random error

### Closed-Form Solution (OLS Estimators)

The OLS estimators that minimize $$\sum_{i=1}^n (y_i - \beta_0 - \beta_1 x_i)^2$$ are:

$$\hat{\beta}_1 = \frac{\sum_{i=1}^{n}(x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^{n}(x_i - \bar{x})^2} = \frac{S_{xy}}{S_{xx}}$$

$$\hat{\beta}_0 = \bar{y} - \hat{\beta}_1 \bar{x}$$

where:
- $$S_{xy} = \sum_{i=1}^n (x_i - \bar{x})(y_i - \bar{y})$$ is the sample covariance numerator
- $$S_{xx} = \sum_{i=1}^n (x_i - \bar{x})^2$$ is the sum of squares of $$x$$

### Relationship to Correlation

$$\hat{\beta}_1 = r_{xy} \cdot \frac{s_y}{s_x}$$

where $$r_{xy}$$ is the Pearson correlation coefficient, $$s_y$$ and $$s_x$$ are sample standard deviations.

### Assumptions of SLR

1. **Linearity**: $$E[y|x] = \beta_0 + \beta_1 x$$ (the true relationship is linear)
2. **Independence**: Observations are independent of each other
3. **Homoscedasticity**: $$\text{Var}(\varepsilon_i) = \sigma^2$$ (constant variance)
4. **Normality**: $$\varepsilon_i \sim \mathcal{N}(0, \sigma^2)$$ (errors are normally distributed)
5. **No measurement error in $$x$$**: The predictor is measured without error

### Variance of Estimators

$$\text{Var}(\hat{\beta}_1) = \frac{\sigma^2}{S_{xx}} = \frac{\sigma^2}{\sum_{i=1}^n(x_i - \bar{x})^2}$$

$$\text{Var}(\hat{\beta}_0) = \sigma^2 \left[\frac{1}{n} + \frac{\bar{x}^2}{S_{xx}}\right]$$

### Interpretation Example

If $$\hat{\beta}_1 = 2.5$$ in a model predicting salary from years of experience:
> *"For each additional year of experience, salary increases by \$2,500 on average, holding all else constant."*

## 4.2 Multiple Linear Regression (MLR)

### Definition

MLR extends SLR to model the relationship between $$p$$ independent variables and a single dependent variable.

### The Model

$$y_i = \beta_0 + \beta_1 x_{i1} + \beta_2 x_{i2} + \cdots + \beta_p x_{ip} + \varepsilon_i$$

Or in matrix form:

$$\mathbf{y} = \mathbf{X}\boldsymbol{\beta} + \boldsymbol{\varepsilon}$$

### Closed-Form Solution (Normal Equations)

The OLS estimator in matrix form:

$$\hat{\boldsymbol{\beta}} = (\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T\mathbf{y}$$

**Derivation**: Setting the gradient of the loss to zero:

$$\frac{\partial}{\partial \boldsymbol{\beta}} \|\mathbf{y} - \mathbf{X}\boldsymbol{\beta}\|^2 = -2\mathbf{X}^T\mathbf{y} + 2\mathbf{X}^T\mathbf{X}\boldsymbol{\beta} = \mathbf{0}$$

$$\Rightarrow \mathbf{X}^T\mathbf{X}\hat{\boldsymbol{\beta}} = \mathbf{X}^T\mathbf{y}$$

$$\Rightarrow \hat{\boldsymbol{\beta}} = (\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T\mathbf{y}$$

*Note*: This requires $$\mathbf{X}^T\mathbf{X}$$ to be invertible (i.e., no perfect multicollinearity).

### The Hat Matrix

The predicted values can be written as:

$$\hat{\mathbf{y}} = \mathbf{X}(\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T\mathbf{y} = \mathbf{H}\mathbf{y}$$

where $$\mathbf{H} = \mathbf{X}(\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T$$ is the **hat matrix** (projection matrix). Properties:
- $$\mathbf{H}$$ is symmetric: $$\mathbf{H}^T = \mathbf{H}$$
- $$\mathbf{H}$$ is idempotent: $$\mathbf{H}^2 = \mathbf{H}$$
- Diagonal elements $$h_{ii}$$ are the **leverage** of each observation

### Assumptions of MLR

All SLR assumptions plus:

6. **No perfect multicollinearity**: No predictor is a perfect linear combination of others. Formally: $$\text{rank}(\mathbf{X}) = p + 1$$
7. **$$n > p + 1$$**: More observations than parameters (for identifiability)

### Gauss-Markov Theorem

Under assumptions 1–6 (linearity, independence, homoscedasticity, zero mean errors, fixed $$\mathbf{X}$$, no perfect multicollinearity) — **but without requiring normality** — the OLS estimator $$\hat{\boldsymbol{\beta}}$$ is the **Best Linear Unbiased Estimator (BLUE)**:

- **Best**: Has the smallest variance among all linear unbiased estimators
- **Linear**: Is a linear function of $$\mathbf{y}$$
- **Unbiased**: $$E[\hat{\boldsymbol{\beta}}] = \boldsymbol{\beta}$$

### Covariance Matrix of $$\hat{\boldsymbol{\beta}}$$

$$\text{Cov}(\hat{\boldsymbol{\beta}}) = \sigma^2 (\mathbf{X}^T\mathbf{X})^{-1}$$

Estimated using $$\hat{\sigma}^2 = \frac{\text{SS}_{\text{res}}}{n - p - 1} = \frac{\sum_{i=1}^n (y_i - \hat{y}_i)^2}{n - p - 1}$$

### Partial Regression Coefficient Interpretation

$$\hat{\beta}_j$$ represents the expected change in $$y$$ for a one-unit increase in $$x_j$$, **holding all other predictors constant** (ceteris paribus). This "holding constant" interpretation is what distinguishes MLR from running separate SLR models.

## 4.3 Polynomial Regression

### Definition

Polynomial regression captures **non-linear** relationships while remaining a *linear model* (linear in parameters, not in features).

### The Model

$$y_i = \beta_0 + \beta_1 x_i + \beta_2 x_i^2 + \beta_3 x_i^3 + \cdots + \beta_d x_i^d + \varepsilon_i$$

for a polynomial of degree $$d$$.

### Why It's Still "Linear" Regression

Define new features: $$z_1 = x, \; z_2 = x^2, \; \ldots, \; z_d = x^d$$. Then:

$$y_i = \beta_0 + \beta_1 z_{i1} + \beta_2 z_{i2} + \cdots + \beta_d z_{id} + \varepsilon_i$$

This is simply MLR in the transformed feature space. All OLS theory applies.

### Key Considerations

| Aspect | Details |
| --- | --- |
| **Degree selection** | Use cross-validation; higher $$d$$ increases flexibility but risks overfitting |
| **Multicollinearity** | Powers of $$x$$ are highly correlated; use orthogonal polynomials or centering |
| **Extrapolation danger** | Polynomials diverge wildly outside the observed range of $$x$$ |
| **Bias-Variance tradeoff** | $$d = 1$$ → high bias, low variance; $$d = n-1$$ → zero bias, infinite variance |

### Assumptions

Same as MLR (the transformation is deterministic), but:
- **Linearity** now means $$E[y|x]$$ is a polynomial of degree $$d$$ (not a straight line)
- Multicollinearity among powers is expected and handled via regularization or orthogonal polynomials

---

## 4.4 Regularized Regression

### Motivation

When $$p$$ is large relative to $$n$$, or predictors are correlated:
- OLS becomes unstable (high variance in $$\hat{\boldsymbol{\beta}}$$)
- $$(\mathbf{X}^T\mathbf{X})^{-1}$$ may not exist or may amplify noise
- The model overfits the training data

**Solution**: Add a penalty term to the loss function to **shrink** coefficients toward zero.

---

### 4.4.1 Ridge Regression ($$L_2$$ Penalty)

**Loss Function**:

$$\mathcal{L}_{\text{Ridge}} = \sum_{i=1}^n (y_i - \hat{y}_i)^2 + \lambda \sum_{j=1}^p \beta_j^2 = \|\mathbf{y} - \mathbf{X}\boldsymbol{\beta}\|^2 + \lambda \|\boldsymbol{\beta}\|_2^2$$

**Closed-Form Solution**:

$$\hat{\boldsymbol{\beta}}_{\text{Ridge}} = (\mathbf{X}^T\mathbf{X} + \lambda \mathbf{I})^{-1}\mathbf{X}^T\mathbf{y}$$

**Properties**:
- Always has a unique solution ($$\mathbf{X}^T\mathbf{X} + \lambda\mathbf{I}$$ is always invertible for $$\lambda > 0$$)
- Shrinks all coefficients toward zero but **never exactly to zero**
- $$\lambda = 0$$ recovers OLS; $$\lambda \to \infty$$ shrinks all $$\hat{\beta}_j \to 0$$
- **Bias-Variance**: Introduces bias but reduces variance (net MSE reduction)
- Handles multicollinearity gracefully

---

### 4.4.2 LASSO Regression ($$L_1$$ Penalty)

**Loss Function**:

$$\mathcal{L}_{\text{LASSO}} = \sum_{i=1}^n (y_i - \hat{y}_i)^2 + \lambda \sum_{j=1}^p |\beta_j| = \|\mathbf{y} - \mathbf{X}\boldsymbol{\beta}\|^2 + \lambda \|\boldsymbol{\beta}\|_1$$

**Properties**:
- **No closed-form solution** (requires iterative optimization: coordinate descent, ISTA)
- Can shrink coefficients **exactly to zero** → performs **automatic feature selection**
- Produces **sparse** models
- Struggles when predictors are highly correlated (selects one arbitrarily)

**Geometric Intuition**: The $$L_1$$ constraint region is a diamond (hypercube corners lie on axes). The OLS solution contours are more likely to intersect the corners, setting some $$\beta_j = 0$$.

---

### 4.4.3 Elastic Net ($$L_1 + L_2$$)

**Loss Function**:

$$\mathcal{L}_{\text{ElasticNet}} = \|\mathbf{y} - \mathbf{X}\boldsymbol{\beta}\|^2 + \lambda_1 \|\boldsymbol{\beta}\|_1 + \lambda_2 \|\boldsymbol{\beta}\|_2^2$$

Often parameterized with mixing ratio $$\alpha \in [0, 1]$$:

$$\mathcal{L} = \|\mathbf{y} - \mathbf{X}\boldsymbol{\beta}\|^2 + \lambda \left[\alpha \|\boldsymbol{\beta}\|_1 + (1 - \alpha)\|\boldsymbol{\beta}\|_2^2\right]$$

where:
- $$\alpha = 1$$ → LASSO
- $$\alpha = 0$$ → Ridge
- $$0 < \alpha < 1$$ → Elastic Net

**Advantages**:
- Performs feature selection (from $$L_1$$) while handling correlated groups (from $$L_2$$)
- Selects **groups of correlated variables** together (grouping effect)
- Numerically more stable than pure LASSO

### Comparison Table

| Property | OLS | Ridge | LASSO | Elastic Net |
| --- | --- | --- | --- | --- |
| Penalty | None | $$\lambda\|\boldsymbol{\beta}\|_2^2$$ | $$\lambda\|\boldsymbol{\beta}\|_1$$ | $$\lambda[\alpha\|\boldsymbol{\beta}\|_1 + (1-\alpha)\|\boldsymbol{\beta}\|_2^2]$$ |
| Feature selection | No | No | Yes | Yes |
| Closed-form | Yes | Yes | No | No |
| Handles multicollinearity | Poorly | Well | Partially | Well |
| Grouping effect | N/A | Yes | No | Yes |
| Sparsity | No | No | Yes | Moderate |

# 5. Assumptions of Linear Regression — A Detailed Treatment

The classical linear regression model (CLM) rests on a set of assumptions often encoded as the **Gauss-Markov conditions** plus normality. Violations have specific consequences and remedies.

---

## 5.1 The LINE Mnemonic

| Letter | Assumption | Formal Statement |
| --- | --- | --- |
| **L** | Linearity | $$E[y \mid \mathbf{x}] = \mathbf{x}^T\boldsymbol{\beta}$$ |
| **I** | Independence | $$\text{Cov}(\varepsilon_i, \varepsilon_j) = 0 \; \forall \; i \neq j$$ |
| **N** | Normality | $$\varepsilon_i \sim \mathcal{N}(0, \sigma^2)$$ |
| **E** | Equal variance (Homoscedasticity) | $$\text{Var}(\varepsilon_i) = \sigma^2 \; \forall \; i$$ |

---

## 5.2 Detailed Breakdown

### Assumption 1: Linearity in Parameters

$$E[y_i \mid x_{i1}, \ldots, x_{ip}] = \beta_0 + \beta_1 x_{i1} + \cdots + \beta_p x_{ip}$$

**What it means**: The conditional mean of $$y$$ is a linear function of the parameters $$\boldsymbol{\beta}$$ (not necessarily linear in $$x$$ — polynomial terms are fine).

**Violation consequence**: Biased and inconsistent parameter estimates.

**Diagnostics**: Residual vs. fitted plot; if curved patterns exist, linearity is violated.

**Remedies**: Transform variables (log, sqrt, polynomial), add interaction terms, or use non-linear models.

---

### Assumption 2: Independence of Errors

$$\text{Cov}(\varepsilon_i, \varepsilon_j) = 0 \quad \text{for all } i \neq j$$

**What it means**: Knowing the error for one observation gives no information about the error for another.

**Violation consequence**: Standard errors are incorrect (usually underestimated) → invalid hypothesis tests and confidence intervals.

**Common violations**: Time series data (autocorrelation), clustered/grouped data.

**Diagnostics**: Durbin-Watson test, autocorrelation function (ACF) plot.

**Remedies**: Generalized Least Squares (GLS), Newey-West standard errors, mixed-effects models.

---

### Assumption 3: Homoscedasticity (Constant Variance)

$$\text{Var}(\varepsilon_i \mid \mathbf{x}_i) = \sigma^2 \quad \text{(constant, does not depend on } \mathbf{x}_i\text{)}$$

**What it means**: The spread of residuals is the same across all levels of the predictors.

**Violation (Heteroscedasticity) consequence**: OLS is still unbiased but **no longer efficient**; standard errors are biased → incorrect inference.

**Diagnostics**: Residual vs. fitted plot (funnel/cone shape), Breusch-Pagan test, White's test.

**Remedies**: Weighted Least Squares (WLS), heteroscedasticity-robust (HC) standard errors (White/Huber-White), log-transform the response.

---

### Assumption 4: Normality of Errors

$$\varepsilon_i \sim \mathcal{N}(0, \sigma^2)$$

**What it means**: Errors follow a Gaussian distribution.

**Why it matters**: Required for exact $$t$$-tests, $$F$$-tests, and confidence intervals in **finite samples**. (By CLT, OLS is asymptotically normal even without this assumption for large $$n$$.)

**Diagnostics**: Q-Q plot, Shapiro-Wilk test, Kolmogorov-Smirnov test.

**Remedies**: Transform $$y$$ (Box-Cox), use bootstrap confidence intervals, or rely on asymptotic theory (large $$n$$).

---

### Assumption 5: No Perfect Multicollinearity

$$\text{rank}(\mathbf{X}) = p + 1$$

**What it means**: No predictor is an exact linear combination of others.

**Violation consequence**: $$(\mathbf{X}^T\mathbf{X})$$ is singular → no unique OLS solution exists.

**Near-multicollinearity**: Coefficients become unstable (huge standard errors), though estimates remain unbiased.

**Diagnostics**: Variance Inflation Factor (VIF):

$$\text{VIF}_j = \frac{1}{1 - R_j^2}$$

where $$R_j^2$$ is the $$R^2$$ from regressing $$x_j$$ on all other predictors. VIF > 10 is commonly flagged.

**Remedies**: Remove correlated predictors, combine via PCA, use Ridge/LASSO regularization.

---

### Assumption 6: Exogeneity (Zero Conditional Mean)

$$E[\varepsilon_i \mid \mathbf{X}] = 0$$

**What it means**: The error is uncorrelated with the predictors. This rules out omitted variable bias, simultaneity, and measurement error in $$x$$.

**Violation consequence**: OLS is **biased and inconsistent**.

**Remedies**: Instrumental variables (2SLS), control variables, natural experiments.

# 6. The Loss Function: Ordinary Least Squares (OLS)

## 6.1 Defining the Objective

The OLS loss function (also called the **residual sum of squares**, RSS or SSE) is:

$$\mathcal{L}(\boldsymbol{\beta}) = \text{RSS} = \sum_{i=1}^n (y_i - \hat{y}_i)^2 = \sum_{i=1}^n \left(y_i - \beta_0 - \sum_{j=1}^p \beta_j x_{ij}\right)^2$$

In matrix form:

$$\mathcal{L}(\boldsymbol{\beta}) = (\mathbf{y} - \mathbf{X}\boldsymbol{\beta})^T(\mathbf{y} - \mathbf{X}\boldsymbol{\beta}) = \|\mathbf{y} - \mathbf{X}\boldsymbol{\beta}\|_2^2$$

## 6.2 Why Squared Errors?

| Reason | Explanation |
| --- | --- |
| **Differentiability** | $$|e|^2$$ is smooth everywhere; $$|e|$$ has a kink at 0 |
| **Penalizes large errors** | Squaring amplifies outlier influence (can be a pro or con) |
| **Connection to MLE** | Under Gaussian errors, minimizing RSS = maximizing likelihood |
| **Unique closed-form** | Quadratic objective → unique global minimum (convex problem) |
| **Geometric elegance** | Minimizes Euclidean distance from $$\mathbf{y}$$ to the column space of $$\mathbf{X}$$ |

## 6.3 Derivation of OLS Estimators (Scalar Case — SLR)

For $$y_i = \beta_0 + \beta_1 x_i + \varepsilon_i$$:

$$\mathcal{L}(\beta_0, \beta_1) = \sum_{i=1}^n (y_i - \beta_0 - \beta_1 x_i)^2$$

**Step 1**: Take partial derivatives and set to zero:

$$\frac{\partial \mathcal{L}}{\partial \beta_0} = -2\sum_{i=1}^n (y_i - \beta_0 - \beta_1 x_i) = 0$$

$$\frac{\partial \mathcal{L}}{\partial \beta_1} = -2\sum_{i=1}^n x_i(y_i - \beta_0 - \beta_1 x_i) = 0$$

**Step 2**: From the first equation:

$$\sum_{i=1}^n y_i - n\beta_0 - \beta_1 \sum_{i=1}^n x_i = 0$$

$$\Rightarrow \hat{\beta}_0 = \bar{y} - \hat{\beta}_1 \bar{x}$$

**Step 3**: Substitute into the second equation and solve:

$$\sum_{i=1}^n x_i(y_i - (\bar{y} - \hat{\beta}_1 \bar{x}) - \hat{\beta}_1 x_i) = 0$$

$$\sum_{i=1}^n x_i(y_i - \bar{y}) - \hat{\beta}_1 \sum_{i=1}^n x_i(x_i - \bar{x}) = 0$$

$$\hat{\beta}_1 = \frac{\sum_{i=1}^n (x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^n (x_i - \bar{x})^2}$$

## 6.4 Derivation of OLS Estimators (Matrix Case — MLR)

Expanding the matrix loss:

$$\mathcal{L}(\boldsymbol{\beta}) = \mathbf{y}^T\mathbf{y} - 2\boldsymbol{\beta}^T\mathbf{X}^T\mathbf{y} + \boldsymbol{\beta}^T\mathbf{X}^T\mathbf{X}\boldsymbol{\beta}$$

Taking the gradient with respect to $$\boldsymbol{\beta}$$:

$$\frac{\partial \mathcal{L}}{\partial \boldsymbol{\beta}} = -2\mathbf{X}^T\mathbf{y} + 2\mathbf{X}^T\mathbf{X}\boldsymbol{\beta}$$

Setting to zero:

$$\mathbf{X}^T\mathbf{X}\hat{\boldsymbol{\beta}} = \mathbf{X}^T\mathbf{y}$$

These are the **Normal Equations**. Solving:

$$\boxed{\hat{\boldsymbol{\beta}} = (\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T\mathbf{y}}$$

**Second-order condition**: The Hessian $$\frac{\partial^2 \mathcal{L}}{\partial \boldsymbol{\beta} \partial \boldsymbol{\beta}^T} = 2\mathbf{X}^T\mathbf{X}$$ is positive semi-definite (positive definite when $$\mathbf{X}$$ has full column rank), confirming this is a minimum.

## 6.5 Alternative Loss Functions (for comparison)

| Loss Function | Formula | Properties |
| --- | --- | --- |
| **Mean Squared Error (MSE)** | $$\frac{1}{n}\sum(y_i - \hat{y}_i)^2$$ | Scaled RSS; same optimum |
| **Mean Absolute Error (MAE)** | $$\frac{1}{n}\sum|y_i - \hat{y}_i|$$ | Robust to outliers; no closed-form |
| **Huber Loss** | Quadratic for small errors, linear for large | Compromise between MSE and MAE |
| **Quantile Loss** | $$\sum \rho_\tau(y_i - \hat{y}_i)$$ | For quantile regression |

# 7. Maximum Likelihood Estimation (MLE) Derivation

This section proves that **under normally distributed errors, the MLE of $$\boldsymbol{\beta}$$ is identical to the OLS estimator**. This is the deep connection Gauss established.

---

## 7.1 Setup: The Probabilistic Model

Assume the linear model with Gaussian errors:

$$y_i = \mathbf{x}_i^T\boldsymbol{\beta} + \varepsilon_i, \quad \varepsilon_i \stackrel{\text{iid}}{\sim} \mathcal{N}(0, \sigma^2)$$

This implies each $$y_i$$ is conditionally normally distributed:

$$y_i \mid \mathbf{x}_i \sim \mathcal{N}(\mathbf{x}_i^T\boldsymbol{\beta}, \; \sigma^2)$$

The probability density function (PDF) of a single observation:

$$f(y_i \mid \mathbf{x}_i; \boldsymbol{\beta}, \sigma^2) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(y_i - \mathbf{x}_i^T\boldsymbol{\beta})^2}{2\sigma^2}\right)$$

---

## 7.2 The Likelihood Function

Since observations are independent, the **joint likelihood** is the product of individual PDFs:

$$L(\boldsymbol{\beta}, \sigma^2 \mid \mathbf{y}, \mathbf{X}) = \prod_{i=1}^n f(y_i \mid \mathbf{x}_i; \boldsymbol{\beta}, \sigma^2)$$

$$L(\boldsymbol{\beta}, \sigma^2) = \prod_{i=1}^n \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(y_i - \mathbf{x}_i^T\boldsymbol{\beta})^2}{2\sigma^2}\right)$$

$$L(\boldsymbol{\beta}, \sigma^2) = \left(\frac{1}{2\pi\sigma^2}\right)^{n/2} \exp\left(-\frac{1}{2\sigma^2}\sum_{i=1}^n (y_i - \mathbf{x}_i^T\boldsymbol{\beta})^2\right)$$

---

## 7.3 The Log-Likelihood Function

Taking the natural logarithm (a monotonic transformation that preserves the maximum):

$$\ell(\boldsymbol{\beta}, \sigma^2) = \ln L(\boldsymbol{\beta}, \sigma^2)$$

$$\ell(\boldsymbol{\beta}, \sigma^2) = -\frac{n}{2}\ln(2\pi) - \frac{n}{2}\ln(\sigma^2) - \frac{1}{2\sigma^2}\sum_{i=1}^n (y_i - \mathbf{x}_i^T\boldsymbol{\beta})^2$$

---

## 7.4 Maximizing w.r.t. $$\boldsymbol{\beta}$$

The terms $$-\frac{n}{2}\ln(2\pi)$$ and $$-\frac{n}{2}\ln(\sigma^2)$$ do not depend on $$\boldsymbol{\beta}$$. Therefore:

$$\hat{\boldsymbol{\beta}}_{\text{MLE}} = \arg\max_{\boldsymbol{\beta}} \; \ell(\boldsymbol{\beta}, \sigma^2) = \arg\min_{\boldsymbol{\beta}} \sum_{i=1}^n (y_i - \mathbf{x}_i^T\boldsymbol{\beta})^2$$

> **Key Result**: Maximizing the log-likelihood with respect to $$\boldsymbol{\beta}$$ is equivalent to minimizing the sum of squared residuals!

Therefore:

$$\boxed{\hat{\boldsymbol{\beta}}_{\text{MLE}} = \hat{\boldsymbol{\beta}}_{\text{OLS}} = (\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T\mathbf{y}}$$

---

## 7.5 Maximizing w.r.t. $$\sigma^2$$

Taking the derivative of the log-likelihood with respect to $$\sigma^2$$:

$$\frac{\partial \ell}{\partial \sigma^2} = -\frac{n}{2\sigma^2} + \frac{1}{2(\sigma^2)^2}\sum_{i=1}^n (y_i - \mathbf{x}_i^T\hat{\boldsymbol{\beta}})^2 = 0$$

Solving:

$$\hat{\sigma}^2_{\text{MLE}} = \frac{1}{n}\sum_{i=1}^n (y_i - \hat{y}_i)^2 = \frac{\text{RSS}}{n}$$

**Important note**: The MLE of $$\sigma^2$$ is **biased** (divides by $$n$$ instead of $$n - p - 1$$). The unbiased estimator is:

$$s^2 = \frac{\text{RSS}}{n - p - 1}$$

---

## 7.6 Summary of the MLE-OLS Connection

| Aspect | Statement |
| --- | --- |
| **Model** | $$y_i \sim \mathcal{N}(\mathbf{x}_i^T\boldsymbol{\beta}, \sigma^2)$$ |
| **MLE of $$\boldsymbol{\beta}$$** | $$(\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T\mathbf{y}$$ = OLS estimator |
| **MLE of $$\sigma^2$$** | $$\text{RSS}/n$$ (biased) |
| **Key insight** | Gaussian errors → maximizing likelihood ≡ minimizing squared errors |
| **Implication** | OLS has the theoretical backing of MLE's optimality properties (consistency, asymptotic efficiency, asymptotic normality) |

## 7.7 What If Errors Aren't Normal?

If $$\varepsilon_i$$ follows a **Laplace distribution** instead, MLE yields **Least Absolute Deviations** (LAD / median regression), not OLS. The choice of loss function is implicitly a distributional assumption about the errors:

| Error Distribution | MLE Loss | Regression Type |
| --- | --- | --- |
| Gaussian ($$\mathcal{N}$$) | Squared loss ($$L_2$$) | OLS |
| Laplace | Absolute loss ($$L_1$$) | LAD / Median regression |
| Student's $$t$$ | Modified squared loss | Robust regression |
| Asymmetric Laplace | Check/quantile loss | Quantile regression |

# 8. Statistical Significance & Hypothesis Testing

## 8.1 The Central Question

After fitting a regression model, we must ask: *Are the estimated relationships real, or could they have arisen by chance?* Statistical testing provides a rigorous framework to answer this.

---

## 8.2 Testing Individual Coefficients: The $$t$$-Test

### Hypotheses

For each coefficient $$\beta_j$$:

$$H_0: \beta_j = 0 \quad \text{(predictor } x_j \text{ has no effect on } y \text{)}$$
$$H_1: \beta_j \neq 0 \quad \text{(predictor } x_j \text{ has a significant effect)}$$

### Test Statistic

$$t_j = \frac{\hat{\beta}_j}{\text{SE}(\hat{\beta}_j)} = \frac{\hat{\beta}_j}{\sqrt{s^2 \cdot [(\mathbf{X}^T\mathbf{X})^{-1}]_{jj}}}$$

Under $$H_0$$, this follows a **Student's $$t$$-distribution** with $$n - p - 1$$ degrees of freedom:

$$t_j \sim t_{n-p-1}$$

### Decision Rule

Reject $$H_0$$ at significance level $$\alpha$$ if:

$$|t_j| > t_{\alpha/2, \; n-p-1}$$

Or equivalently, if the p-value satisfies:

$$\text{p-value} = 2 \cdot P(T > |t_j|) < \alpha$$

### Interpretation

- **Large $$|t_j|$$** (or small p-value) → strong evidence that $$\beta_j \neq 0$$
- A non-significant $$\beta_j$$ does **not** prove $$x_j$$ is irrelevant (could be underpowered, or $$x_j$$'s effect is absorbed by correlated predictors)

---

## 8.3 Confidence Intervals for Coefficients

A $$(1-\alpha) \times 100\%$$ confidence interval for $$\beta_j$$:

$$\hat{\beta}_j \pm t_{\alpha/2, \; n-p-1} \cdot \text{SE}(\hat{\beta}_j)$$

**Interpretation**: If we repeated the experiment many times, $$(1-\alpha)\%$$ of such intervals would contain the true $$\beta_j$$.

If the CI **excludes zero**, the coefficient is statistically significant at level $$\alpha$$.

---

## 8.4 Testing Overall Model Significance: The $$F$$-Test

### Hypotheses

$$H_0: \beta_1 = \beta_2 = \cdots = \beta_p = 0 \quad \text{(no predictor matters)}$$
$$H_1: \text{At least one } \beta_j \neq 0$$

### Test Statistic

$$F = \frac{(\text{SS}_{\text{tot}} - \text{SS}_{\text{res}}) / p}{\text{SS}_{\text{res}} / (n - p - 1)} = \frac{\text{SS}_{\text{reg}} / p}{\text{SS}_{\text{res}} / (n - p - 1)} = \frac{\text{MSR}}{\text{MSE}}$$

where:
- $$\text{SS}_{\text{reg}} = \sum(\hat{y}_i - \bar{y})^2$$ = regression sum of squares
- $$\text{MSR} = \text{SS}_{\text{reg}}/p$$ = mean square regression
- $$\text{MSE} = \text{SS}_{\text{res}}/(n-p-1)$$ = mean square error

Under $$H_0$$:

$$F \sim F_{p, \; n-p-1}$$

### ANOVA Table

| Source | df | Sum of Squares | Mean Square | F |
| --- | --- | --- | --- | --- |
| Regression | $$p$$ | $$\text{SS}_{\text{reg}}$$ | $$\text{MSR}$$ | $$\text{MSR}/\text{MSE}$$ |
| Residual | $$n-p-1$$ | $$\text{SS}_{\text{res}}$$ | $$\text{MSE}$$ | |
| Total | $$n-1$$ | $$\text{SS}_{\text{tot}}$$ | | |

---

## 8.5 Partial $$F$$-Test (Testing Subsets of Coefficients)

To test whether a subset of $$q$$ predictors contributes significantly:

$$H_0: \beta_{p-q+1} = \cdots = \beta_p = 0$$

$$F = \frac{(\text{RSS}_{\text{reduced}} - \text{RSS}_{\text{full}}) / q}{\text{RSS}_{\text{full}} / (n - p - 1)} \sim F_{q, \; n-p-1}$$

where $$\text{RSS}_{\text{reduced}}$$ is the RSS from the model without those $$q$$ predictors.

---

## 8.6 p-Value Interpretation

| p-value | Evidence Against $$H_0$$ |
| --- | --- |
| $$> 0.10$$ | No evidence |
| $$0.05 - 0.10$$ | Weak (marginal) evidence |
| $$0.01 - 0.05$$ | Moderate evidence |
| $$0.001 - 0.01$$ | Strong evidence |
| $$< 0.001$$ | Very strong evidence |

**Caution**: Statistical significance $$\neq$$ practical significance. A coefficient can be statistically significant but economically/practically trivial (especially with large $$n$$).

---

## 8.7 Multiple Testing Problem

When testing many coefficients simultaneously, the probability of at least one false positive (Type I error) increases:

$$P(\text{at least one false positive}) = 1 - (1 - \alpha)^m$$

For $$m = 20$$ tests at $$\alpha = 0.05$$: $$P \approx 0.64$$ (64% chance of a false positive!).

**Corrections**:
- **Bonferroni**: Use $$\alpha/m$$ for each test (conservative)
- **Benjamini-Hochberg (FDR)**: Controls the false discovery rate
- **Holm-Bonferroni**: Step-down procedure (less conservative than Bonferroni)

# 9. Model Evaluation & Goodness of Fit

## 9.1 Coefficient of Determination ($$R^2$$)

$$R^2 = 1 - \frac{\text{SS}_{\text{res}}}{\text{SS}_{\text{tot}}} = 1 - \frac{\sum(y_i - \hat{y}_i)^2}{\sum(y_i - \bar{y})^2}$$

**Properties**:
- $$0 \leq R^2 \leq 1$$ (for models with an intercept)
- $$R^2 = 0$$: Model explains no variance beyond the mean
- $$R^2 = 1$$: Perfect fit (all points lie on the regression surface)
- **Always increases** when adding predictors (even irrelevant ones)
- In SLR: $$R^2 = r_{xy}^2$$ (squared Pearson correlation)

**Decomposition of Variance**:

$$\underbrace{\text{SS}_{\text{tot}}}_{\text{Total variation}} = \underbrace{\text{SS}_{\text{reg}}}_{\text{Explained variation}} + \underbrace{\text{SS}_{\text{res}}}_{\text{Unexplained variation}}$$

---

## 9.2 Adjusted $$R^2$$

Penalizes for adding predictors that don't improve the model:

$$R^2_{\text{adj}} = 1 - \frac{\text{SS}_{\text{res}} / (n - p - 1)}{\text{SS}_{\text{tot}} / (n - 1)} = 1 - \frac{(1 - R^2)(n - 1)}{n - p - 1}$$

**Properties**:
- Can decrease if a new predictor doesn't sufficiently reduce RSS
- Can be negative (if the model is worse than the mean)
- Better for comparing models with different numbers of predictors
- Relationship: $$R^2_{\text{adj}} = 1 - \frac{\text{MSE}}{\text{Var}(y)}$$

---

## 9.3 Information Criteria

### Akaike Information Criterion (AIC)

$$\text{AIC} = n \ln\left(\frac{\text{RSS}}{n}\right) + 2(p + 1)$$

### Bayesian Information Criterion (BIC)

$$\text{BIC} = n \ln\left(\frac{\text{RSS}}{n}\right) + (p + 1)\ln(n)$$

| Criterion | Penalty | Preference |
| --- | --- | --- |
| AIC | $$2k$$ (lighter) | Prediction-oriented; may overfit with many predictors |
| BIC | $$k \ln(n)$$ (heavier for $$n > 7$$) | Consistent (selects true model as $$n \to \infty$$) |

*Lower values = better model*

---

## 9.4 Residual Analysis

### Types of Residuals

| Type | Formula | Use |
| --- | --- | --- |
| Raw | $$e_i = y_i - \hat{y}_i$$ | Basic diagnostic |
| Standardized | $$r_i = e_i / s$$ | Compare residual magnitudes |
| Studentized (internally) | $$r_i = e_i / (s\sqrt{1 - h_{ii}})$$ | Account for leverage |
| Studentized (externally/deleted) | $$t_i = e_i / (s_{(i)}\sqrt{1 - h_{ii}})$$ | Most reliable for outlier detection |

### Diagnostic Plots

1. **Residuals vs. Fitted Values**: Check linearity and homoscedasticity
2. **Normal Q-Q Plot**: Check normality of residuals
3. **Scale-Location Plot** ($$\sqrt{|\text{standardized residuals}|}$$ vs. fitted): Check homoscedasticity
4. **Residuals vs. Leverage**: Identify influential observations (Cook's distance contours)

---

## 9.5 Influence Measures

### Leverage ($$h_{ii}$$)

$$h_{ii} = [\mathbf{H}]_{ii} = \mathbf{x}_i^T(\mathbf{X}^T\mathbf{X})^{-1}\mathbf{x}_i$$

- Measures how far $$\mathbf{x}_i$$ is from the center of the predictor space
- $$1/n \leq h_{ii} \leq 1$$; sum is $$p + 1$$
- Rule of thumb: $$h_{ii} > 2(p+1)/n$$ is high leverage

### Cook's Distance

$$D_i = \frac{(\hat{\mathbf{y}} - \hat{\mathbf{y}}_{(i)})^T(\hat{\mathbf{y}} - \hat{\mathbf{y}}_{(i)})}{(p+1) \cdot \text{MSE}} = \frac{r_i^2}{p+1} \cdot \frac{h_{ii}}{1 - h_{ii}}$$

- Combines leverage and residual magnitude
- Measures overall influence of observation $$i$$ on all fitted values
- Rule of thumb: $$D_i > 1$$ (or $$> 4/n$$) indicates an influential point

### DFFITS

$$\text{DFFITS}_i = \frac{\hat{y}_i - \hat{y}_{(i)}}{s_{(i)}\sqrt{h_{ii}}}$$

- Measures influence on individual fitted value
- Threshold: $$|\text{DFFITS}_i| > 2\sqrt{(p+1)/n}$$

# 10. Interpretability Techniques

Linear regression is often called the "gold standard" of interpretable models. Here we systematize the tools available for interpreting regression outputs.

---

## 10.1 Direct Coefficient Interpretation

### Unstandardized Coefficients

$$\hat{\beta}_j$$: the expected change in $$y$$ for a one-unit increase in $$x_j$$, holding all other predictors constant.

**Limitations**: Cannot compare coefficients across predictors with different scales.

### Standardized Coefficients (Beta Weights)

Standardize both $$x$$ and $$y$$ before fitting:

$$\hat{\beta}_j^{\text{std}} = \hat{\beta}_j \cdot \frac{s_{x_j}}{s_y}$$

**Interpretation**: A one-standard-deviation increase in $$x_j$$ is associated with a $$\hat{\beta}_j^{\text{std}}$$ standard deviation change in $$y$$.

**Advantage**: Comparable across predictors regardless of measurement units.

---

## 10.2 Marginal Effects and Elasticities

### Marginal Effect (Linear Model)

In a purely linear model, the marginal effect of $$x_j$$ is constant:

$$\frac{\partial E[y]}{\partial x_j} = \beta_j$$

### Elasticity

For log-transformed models:

| Model | Equation | Interpretation of $$\beta_1$$ |
| --- | --- | --- |
| Level-Level | $$y = \beta_0 + \beta_1 x$$ | 1-unit ↑ in $$x$$ → $$\beta_1$$ units ↑ in $$y$$ |
| Log-Level | $$\ln(y) = \beta_0 + \beta_1 x$$ | 1-unit ↑ in $$x$$ → $$100\beta_1$$% ↑ in $$y$$ |
| Level-Log | $$y = \beta_0 + \beta_1 \ln(x)$$ | 1% ↑ in $$x$$ → $$\beta_1/100$$ units ↑ in $$y$$ |
| Log-Log | $$\ln(y) = \beta_0 + \beta_1 \ln(x)$$ | 1% ↑ in $$x$$ → $$\beta_1$$% ↑ in $$y$$ (elasticity) |

---

## 10.3 Partial Dependence and Ceteris Paribus

### Added Variable Plot (Partial Regression Plot)

To visualize the **unique contribution** of $$x_j$$ after removing the effect of all other predictors:

1. Regress $$y$$ on all predictors *except* $$x_j$$ → residuals $$e_y$$
2. Regress $$x_j$$ on all other predictors → residuals $$e_{x_j}$$
3. Plot $$e_y$$ vs. $$e_{x_j}$$

The slope of this plot equals $$\hat{\beta}_j$$ from the full model.

---

## 10.4 Variable Importance

### Method 1: Absolute Standardized Coefficients

$$\text{Importance}(x_j) = |\hat{\beta}_j^{\text{std}}|$$

### Method 2: Partial $$F$$-test (Type III SS)

Measures the unique contribution of each predictor after all others:

$$\Delta R^2_j = R^2_{\text{full}} - R^2_{\text{without } x_j}$$

### Method 3: Dominance Analysis

Averages the $$\Delta R^2$$ contribution of $$x_j$$ across **all possible subset models** — provides the most comprehensive measure but is computationally expensive ($$2^p$$ models).

### Method 4: SHAP Values (SHapley Additive exPlanations)

For linear regression, the SHAP value of feature $$j$$ for observation $$i$$ is:

$$\phi_j^{(i)} = \hat{\beta}_j (x_{ij} - \bar{x}_j)$$

The sum of all SHAP values equals the difference between the prediction and the mean:

$$\hat{y}_i - \bar{y} = \sum_{j=1}^p \phi_j^{(i)}$$

---

## 10.5 Confidence vs. Prediction Intervals

### Confidence Interval (for the mean response)

The interval for $$E[y \mid \mathbf{x}_0]$$:

$$\hat{y}_0 \pm t_{\alpha/2, \; n-p-1} \cdot s \sqrt{\mathbf{x}_0^T(\mathbf{X}^T\mathbf{X})^{-1}\mathbf{x}_0}$$

### Prediction Interval (for a new observation)

The interval for a single new $$y_0$$:

$$\hat{y}_0 \pm t_{\alpha/2, \; n-p-1} \cdot s \sqrt{1 + \mathbf{x}_0^T(\mathbf{X}^T\mathbf{X})^{-1}\mathbf{x}_0}$$

**Key difference**: The prediction interval is always wider because it includes both estimation uncertainty *and* irreducible error $$\sigma^2$$.

---

## 10.6 Interaction Effects

When two predictors interact:

$$y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \beta_3 (x_1 \cdot x_2) + \varepsilon$$

The marginal effect of $$x_1$$ now **depends on the value of $$x_2$$**:

$$\frac{\partial E[y]}{\partial x_1} = \beta_1 + \beta_3 x_2$$

**Interpretation**: $$\beta_3$$ measures how the effect of $$x_1$$ on $$y$$ *changes* as $$x_2$$ varies.

---

## 10.7 Decomposition of Predictions

For any prediction $$\hat{y}_i$$, we can decompose it into individual predictor contributions:

$$\hat{y}_i = \hat{\beta}_0 + \underbrace{\hat{\beta}_1 x_{i1}}_{\text{contribution of } x_1} + \underbrace{\hat{\beta}_2 x_{i2}}_{\text{contribution of } x_2} + \cdots + \underbrace{\hat{\beta}_p x_{ip}}_{\text{contribution of } x_p}$$

This additive decomposition is a hallmark of linear regression's interpretability — each predictor's contribution is **transparent, additive, and independent** of the values of other predictors.

# 12. Tough Interview Questions & Answers

---

## Q1. If OLS is BLUE (Best Linear Unbiased Estimator), why would anyone ever use a biased estimator like Ridge?

**Answer:**

BLUE means OLS has the **lowest variance among all *unbiased* linear estimators**. But the Gauss-Markov theorem says nothing about **Mean Squared Error (MSE)**, which decomposes as:

$$\text{MSE}(\hat{\beta}) = \text{Bias}^2(\hat{\beta}) + \text{Var}(\hat{\beta})$$

Ridge introduces a small bias ($$\text{Bias} = -\lambda(\mathbf{X}^T\mathbf{X} + \lambda\mathbf{I})^{-1}\boldsymbol{\beta}$$) but can dramatically reduce variance, especially when predictors are correlated. The **net MSE can be lower** than OLS — making Ridge superior in prediction despite being biased. This is the classic bias-variance tradeoff: a little bias buys a lot of variance reduction.

---

## Q2. You fit a regression and get $$R^2 = 0.95$$. Your manager says the model is excellent. What's your concern?

**Answer:**

High $$R^2$$ alone is **not** sufficient evidence of a good model. Concerns include:

1. **Overfitting**: With many predictors, $$R^2$$ always increases. Check Adjusted $$R^2$$ or out-of-sample performance.
2. **Spurious correlation**: Two trending time series (e.g., GDP and ice cream sales) will have high $$R^2$$ without causal meaning.
3. **Residual structure**: If residuals show patterns (autocorrelation, heteroscedasticity), the model is misspecified despite high $$R^2$$.
4. **Omitted variable bias**: A high $$R^2$$ model can still produce biased coefficient estimates if key confounders are missing.
5. **Data leakage**: Features that encode the target (e.g., `total_price` predicting `revenue`) inflate $$R^2$$ artificially.

*Always pair $$R^2$$ with residual diagnostics, cross-validation, and domain sense.*

---

## Q3. Can $$R^2$$ be negative? When would that happen?

**Answer:**

Yes, $$R^2$$ can be negative when the model is evaluated on **test data** (or when no intercept is fitted).

$$R^2 = 1 - \frac{\sum(y_i - \hat{y}_i)^2}{\sum(y_i - \bar{y})^2}$$

If $$\text{SS}_{\text{res}} > \text{SS}_{\text{tot}}$$, i.e., the model's predictions are **worse than simply predicting the mean**, then $$R^2 < 0$$. This happens when:
- The model severely overfits training data and fails on test data
- A regression through the origin is forced when the data doesn't warrant it
- Predictions from one dataset are evaluated against a differently-distributed dataset

*Note*: On **training data with an intercept**, $$R^2 \geq 0$$ is guaranteed.

---

## Q4. You have two models: Model A with $$R^2 = 0.80$$ (3 predictors) and Model B with $$R^2 = 0.82$$ (15 predictors). Which do you prefer?

**Answer:**

Almost certainly **Model A**. The criteria:

1. **Adjusted $$R^2$$**: Penalizes extra parameters. Likely $$R^2_{\text{adj}}(A) > R^2_{\text{adj}}(B)$$ since 12 extra predictors only added 0.02 to $$R^2$$.
2. **Parsimony (Occam's Razor)**: Simpler models generalize better, are more interpretable, and less prone to overfitting.
3. **AIC/BIC**: Both will heavily penalize Model B for 12 extra parameters.
4. **Cross-validation MSE**: The true test — compare out-of-sample prediction error.
5. **Practical**: Fewer features = cheaper data collection, fewer failure points, easier maintenance.

The marginal gain of 0.02 in $$R^2$$ from 12 extra features is almost certainly noise or multicollinearity.

---

## Q5. Explain why multicollinearity inflates standard errors but doesn't bias the coefficients.

**Answer:**

The covariance matrix of $$\hat{\boldsymbol{\beta}}$$ is:

$$\text{Cov}(\hat{\boldsymbol{\beta}}) = \sigma^2(\mathbf{X}^T\mathbf{X})^{-1}$$

When columns of $$\mathbf{X}$$ are nearly collinear, $$(\mathbf{X}^T\mathbf{X})$$ becomes nearly singular → its inverse has very large elements → **variances (diagonal) and covariances blow up**.

However, the estimator $$\hat{\boldsymbol{\beta}} = (\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T\mathbf{y}$$ remains **unbiased** because:

$$E[\hat{\boldsymbol{\beta}}] = (\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T E[\mathbf{y}] = (\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T\mathbf{X}\boldsymbol{\beta} = \boldsymbol{\beta}$$

This derivation holds regardless of the condition number of $$\mathbf{X}^T\mathbf{X}$$. So the estimate is "correct on average" but wildly unstable (small data perturbations cause large coefficient swings).

**Analogy**: Trying to separate the effects of two nearly identical variables is like asking "how much did the left engine vs. right engine contribute to the plane's speed?" when both always run at the same power.

---

## Q6. A feature has a very high p-value (p = 0.85) in your model. Should you always remove it?

**Answer:**

**Not necessarily.** Reasons to keep a non-significant predictor:

1. **Confounding control**: Even if $$x_j$$ isn't significant, its inclusion may be necessary to get unbiased estimates of other coefficients (e.g., controlling for age in a medical study).
2. **Multicollinearity**: $$x_j$$ may be collinear with another significant predictor, inflating its p-value. Removing it could destabilize other coefficients.
3. **Theoretical importance**: Domain knowledge may dictate inclusion regardless of statistical significance.
4. **Interaction effects**: $$x_j$$ alone may not be significant, but $$x_j \cdot x_k$$ could be (hierarchical principle: keep main effects if interactions are significant).
5. **Low power**: With small $$n$$, the test may simply lack power to detect a real but small effect.

*Automated stepwise removal based purely on p-values is generally discouraged in modern practice.*

---

## Q7. Derive why the OLS residuals always sum to zero (when an intercept is included).

**Answer:**

From the first-order condition (Normal Equation) for $$\beta_0$$:

$$\frac{\partial}{\partial \beta_0} \sum_{i=1}^n (y_i - \beta_0 - \beta_1 x_{i1} - \cdots)^2 = 0$$

$$-2\sum_{i=1}^n (y_i - \hat{y}_i) = 0$$

$$\therefore \sum_{i=1}^n e_i = 0$$

In matrix terms, the first column of $$\mathbf{X}$$ is $$\mathbf{1}_n$$ (the intercept column). The orthogonality condition $$\mathbf{X}^T\mathbf{e} = \mathbf{0}$$ implies:

$$\mathbf{1}_n^T \mathbf{e} = \sum_{i=1}^n e_i = 0$$

This is why $$\bar{e} = 0$$ always holds with an intercept, and consequently $$\bar{\hat{y}} = \bar{y}$$.

**Implication**: Without an intercept, residuals need NOT sum to zero, and $$R^2$$ can become negative.

---

## Q8. What happens to OLS when $$n < p$$ (more features than observations)?

**Answer:**

When $$n < p + 1$$:

1. $$\mathbf{X}^T\mathbf{X}$$ is a $$(p+1) \times (p+1)$$ matrix of rank at most $$n < p+1$$ → **singular** (non-invertible)
2. The normal equations $$\mathbf{X}^T\mathbf{X}\hat{\boldsymbol{\beta}} = \mathbf{X}^T\mathbf{y}$$ have **infinitely many solutions**
3. The model can achieve $$\text{RSS} = 0$$ (perfectly interpolate all $$n$$ points) but with no unique $$\hat{\boldsymbol{\beta}}$$
4. The model has **zero generalization ability** — it has memorized the training data

**Solutions**:
- Ridge regression: $$(\mathbf{X}^T\mathbf{X} + \lambda\mathbf{I})$$ is always invertible
- LASSO / Elastic Net: Enforce sparsity, selecting at most $$n$$ non-zero coefficients
- PCA regression: Reduce dimensionality first
- Minimum-norm OLS: $$\hat{\boldsymbol{\beta}} = \mathbf{X}^T(\mathbf{X}\mathbf{X}^T)^{-1}\mathbf{y}$$ (the Moore-Penrose pseudoinverse solution)

## Q9. A colleague says "My regression coefficient for advertising spend is 0.5, so advertising causes 0.5 units increase in sales per dollar spent." What's wrong?

**Answer:**

**Correlation ≠ Causation.** Regression coefficients measure **association**, not causation. For a causal interpretation, you need:

1. **No omitted variable bias**: All confounders must be controlled. E.g., perhaps stores that advertise more are already in high-demand areas.
2. **No reverse causality**: Perhaps higher sales → more budget → more ad spend (simultaneity).
3. **No selection bias**: The data must represent the population of interest.
4. **Exogeneity**: $$E[\varepsilon | \text{ad\_spend}] = 0$$ — the error must be uncorrelated with the predictor.

To establish causality, you need:
- Randomized controlled experiments (A/B tests)
- Instrumental variables (2SLS)
- Regression discontinuity designs
- Difference-in-differences
- Directed Acyclic Graphs (DAGs) with the backdoor criterion

## Q10. Why does LASSO produce exactly zero coefficients but Ridge does not? Explain geometrically.

**Answer:**

The constrained optimization form of regularized regression is:

$$\min_{\boldsymbol{\beta}} \|\mathbf{y} - \mathbf{X}\boldsymbol{\beta}\|^2 \quad \text{subject to} \quad \|\boldsymbol{\beta}\|_q \leq t$$

**Geometric picture** (2D case with $$\beta_1, \beta_2$$):

- The **OLS loss** produces elliptical contours centered at $$\hat{\boldsymbol{\beta}}_{\text{OLS}}$$
- **Ridge** ($$q = 2$$): Constraint region is a **circle** ($$\beta_1^2 + \beta_2^2 \leq t$$). Elliptical contours typically touch the circle at a point where **neither coordinate is zero**.
- **LASSO** ($$q = 1$$): Constraint region is a **diamond** ($$|\beta_1| + |\beta_2| \leq t$$). The diamond has **corners on the axes**. Elliptical contours are much more likely to first touch the diamond at a corner, where one coordinate **is exactly zero**.

**Mathematically**: The $$L_1$$ penalty's subgradient includes zero in its subdifferential at $$\beta_j = 0$$, creating a finite region where the gradient of the loss can be "absorbed" without moving $$\beta_j$$ away from zero. The $$L_2$$ penalty's gradient at $$\beta_j = 0$$ is also zero, providing no finite "holding force."

## Q11. You run a regression and the Durbin-Watson statistic is 0.8. What does this mean and what should you do?

**Answer:**

The **Durbin-Watson (DW)** statistic tests for first-order autocorrelation in residuals:

$$DW = \frac{\sum_{i=2}^n (e_i - e_{i-1})^2}{\sum_{i=1}^n e_i^2} \approx 2(1 - \hat{\rho})$$

where $$\hat{\rho}$$ is the sample autocorrelation of residuals.

- $$DW \approx 2$$: No autocorrelation
- $$DW \approx 0$$: Strong positive autocorrelation ($$\hat{\rho} \approx 1$$)
- $$DW \approx 4$$: Strong negative autocorrelation

**DW = 0.8** implies $$\hat{\rho} \approx 0.6$$ — strong **positive autocorrelation**. This means:
- The **independence assumption is violated**
- Standard errors are **underestimated** → t-statistics inflated → false significance
- Coefficients remain unbiased but are **inefficient**

**Remedies**:
1. Add lagged dependent variable: $$y_t = \beta_0 + \beta_1 x_t + \phi y_{t-1} + \varepsilon_t$$
2. Use Generalized Least Squares (GLS) with AR(1) error structure
3. Use Newey-West HAC standard errors (heteroscedasticity and autocorrelation consistent)
4. Re-examine model specification — autocorrelation may signal omitted time-varying variables

## Q12. Explain the difference between $$R^2$$ and Adjusted $$R^2$$. Can Adjusted $$R^2$$ decrease when you add a variable?

**Answer:**

$$R^2 = 1 - \frac{\text{SS}_{\text{res}}}{\text{SS}_{\text{tot}}}$$ **always increases** (or stays the same) when a new predictor is added, because adding a variable can only reduce or maintain RSS (the optimization has more freedom).

$$R^2_{\text{adj}} = 1 - \frac{\text{SS}_{\text{res}}/(n-p-1)}{\text{SS}_{\text{tot}}/(n-1)}$$

Adjusted $$R^2$$ **can decrease** because when a new variable is added:
- Numerator: $$\text{SS}_{\text{res}}$$ decreases, but we divide by $$(n-p-2)$$ instead of $$(n-p-1)$$
- If the reduction in RSS is not large enough to compensate for the loss of one degree of freedom, Adjusted $$R^2$$ decreases

**Threshold**: Adding variable $$x_{p+1}$$ increases Adjusted $$R^2$$ if and only if its partial F-statistic exceeds 1:

$$F_{\text{partial}} = \frac{(\text{RSS}_{\text{old}} - \text{RSS}_{\text{new}})/1}{\text{RSS}_{\text{new}}/(n-p-2)} > 1$$

## Q13. In the MLE derivation, why do we take the log of the likelihood? Is it just for convenience?

**Answer:**

Taking the logarithm is justified by **multiple reasons**:

1. **Monotonicity**: $$\ln(\cdot)$$ is monotonically increasing, so $$\arg\max L(\theta) = \arg\max \ln L(\theta)$$. The maximizer is preserved.

2. **Products become sums**: For independent observations, $$L = \prod f_i$$ becomes $$\ell = \sum \ln f_i$$. Sums are:
   - Easier to differentiate
   - Numerically stable (products of small probabilities underflow to zero)

3. **Exponential families simplify**: For Gaussians, the log removes the $$\exp$$, leaving a quadratic in $$\boldsymbol{\beta}$$.

4. **Asymptotic theory**: The log-likelihood ratio has a known $$\chi^2$$ distribution (Wilks' theorem), enabling hypothesis tests.

5. **Information theory**: $$-\ell/n$$ converges to the KL-divergence between the true distribution and the model, giving MLE a deeper information-theoretic meaning.

So yes, it's partly convenience, but it's also **mathematically natural** — the log-likelihood is the fundamental object in statistical theory.

## Q14. You standardize all features before fitting Ridge regression. Why is this important? What happens if you don't?

**Answer:**

The Ridge penalty is:

$$\lambda \sum_{j=1}^p \beta_j^2$$

This penalizes all coefficients **equally in magnitude**. But if features have different scales:
- $$x_1$$ in [0, 1] (e.g., proportion) → $$\beta_1$$ is large to compensate
- $$x_2$$ in [0, 1000000] (e.g., revenue in dollars) → $$\beta_2$$ is tiny

Without standardization, Ridge will **penalize large-scale features less** and **penalize small-scale features more** — purely because of the arbitrary choice of units, not because of actual importance.

**After standardization** ($$z_j = (x_j - \bar{x}_j)/s_j$$): all features are on the same scale, so the penalty is **equitable**. The shrinkage reflects the actual predictive contribution, not measurement units.

*Note*: OLS coefficients are scale-equivariant (changing units in $$x$$ inversely changes $$\beta$$), so standardization doesn't affect OLS predictions — only regularized methods.

## Q15. A model has a significant F-test (overall model significant) but NO individual t-test is significant. How is this possible?

**Answer:**

This is a classic symptom of **multicollinearity**. Here's why:

- The **F-test** asks: "Do the predictors *collectively* explain significant variance?" It tests $$H_0: \beta_1 = \beta_2 = \cdots = \beta_p = 0$$ simultaneously.
- Each **t-test** asks: "Does $$x_j$$ contribute *uniquely*, after accounting for all other predictors?" It tests $$H_0: \beta_j = 0 \mid \text{all others in model}$$.

When predictors are highly correlated:
- Together they explain $$y$$ well (significant F)
- But individually, none can be distinguished as the "unique contributor" because their effects overlap
- The shared variance can't be attributed to any single predictor → all individual SEs inflate → all t-statistics shrink

**Example**: Predicting house price from `square_feet` and `number_of_rooms` (highly correlated). Together they predict well; individually, neither adds much beyond the other.

**Resolution**: Check VIFs, remove or combine correlated predictors, use Ridge/PCA.

## Q16. Prove that in SLR, the regression line always passes through the point $$(\bar{x}, \bar{y})$$.

**Answer:**

From the OLS estimator for the intercept:

$$\hat{\beta}_0 = \bar{y} - \hat{\beta}_1 \bar{x}$$

The predicted value at $$x = \bar{x}$$ is:

$$\hat{y}\big|_{x=\bar{x}} = \hat{\beta}_0 + \hat{\beta}_1 \bar{x} = (\bar{y} - \hat{\beta}_1 \bar{x}) + \hat{\beta}_1 \bar{x} = \bar{y}$$

Therefore, the point $$(\bar{x}, \bar{y})$$ always satisfies the regression equation. $$\blacksquare$$

**Geometric interpretation**: The regression line is "anchored" at the centroid of the data and rotated by the slope $$\hat{\beta}_1$$.

## Q17. What is the difference between "error" ($$\varepsilon$$) and "residual" ($$e$$)? Why does this distinction matter?

**Answer:**

| | Error ($$\varepsilon_i$$) | Residual ($$e_i$$) |
| --- | --- | --- |
| **Definition** | $$\varepsilon_i = y_i - (\beta_0 + \beta_1 x_i)$$ | $$e_i = y_i - (\hat{\beta}_0 + \hat{\beta}_1 x_i)$$ |
| **Uses true parameters** | Yes ($$\boldsymbol{\beta}$$) | No ($$\hat{\boldsymbol{\beta}}$$) |
| **Observable?** | No (never known) | Yes (computed from data) |
| **Properties** | iid $$\mathcal{N}(0, \sigma^2)$$ | NOT independent; $$\text{Cov}(\mathbf{e}) = \sigma^2(\mathbf{I} - \mathbf{H})$$ |
| **Sum** | Not necessarily zero | Always zero (with intercept) |

**Why it matters**:
- We use residuals to **estimate** properties of errors (normality, constant variance)
- But residuals are **correlated** and have **unequal variance** $$\text{Var}(e_i) = \sigma^2(1 - h_{ii})$$, even if true errors are homoscedastic
- This is why **studentized residuals** (dividing by $$s\sqrt{1-h_{ii}}$$) are preferred for diagnostics

## Q18. Explain the "Frisch-Waugh-Lovell" theorem and why it's important.

**Answer:**

The **Frisch-Waugh-Lovell (FWL) Theorem** states:

> In the regression $$y = X_1\beta_1 + X_2\beta_2 + \varepsilon$$, the OLS estimate $$\hat{\beta}_2$$ is identical to the coefficient obtained from regressing the residuals of $$y$$ on $$X_1$$ against the residuals of $$X_2$$ on $$X_1$$.

Formally:
1. Regress $$\mathbf{y}$$ on $$\mathbf{X}_1$$ → get residuals $$\mathbf{M}_1 \mathbf{y}$$
2. Regress each column of $$\mathbf{X}_2$$ on $$\mathbf{X}_1$$ → get residuals $$\mathbf{M}_1 \mathbf{X}_2$$
3. Regress $$\mathbf{M}_1 \mathbf{y}$$ on $$\mathbf{M}_1 \mathbf{X}_2$$ → this yields exactly $$\hat{\beta}_2$$

where $$\mathbf{M}_1 = \mathbf{I} - \mathbf{X}_1(\mathbf{X}_1^T\mathbf{X}_1)^{-1}\mathbf{X}_1^T$$ is the residual-maker matrix.

**Why it's important**:
- Provides the mathematical foundation for "controlling for" variables
- Explains the added variable plot (partial regression plot)
- Justifies the "partialling out" interpretation of multiple regression coefficients
- Computationally useful: estimate one coefficient without re-running the full regression
- Underpins many econometric methods (fixed effects = partialling out group means)

## Q19. You fit a regression with $$\log(y)$$ as the response. How do you correctly predict $$y$$ (not $$\log y$$) for new data? Why can't you just exponentiate $$\hat{\mu}$$?

**Answer:**

If the model is $$\ln(y) = \mathbf{x}^T\boldsymbol{\beta} + \varepsilon$$ with $$\varepsilon \sim \mathcal{N}(0, \sigma^2)$$, then:

$$y = e^{\mathbf{x}^T\boldsymbol{\beta} + \varepsilon} = e^{\mathbf{x}^T\boldsymbol{\beta}} \cdot e^{\varepsilon}$$

The **expected value** of $$y$$ is:

$$E[y] = e^{\mathbf{x}^T\boldsymbol{\beta}} \cdot E[e^{\varepsilon}] = e^{\mathbf{x}^T\boldsymbol{\beta}} \cdot e^{\sigma^2/2}$$

(using the moment generating function of the normal distribution)

Simply exponentiating $$\hat{\mu} = \mathbf{x}^T\hat{\boldsymbol{\beta}}$$ gives the **median** of $$y$$ (since $$\exp$$ is monotone and the median of $$\ln y$$ maps to the median of $$y$$), but **not the mean**.

The correct retransformation is:

$$\hat{E}[y] = \exp\left(\hat{\mu} + \frac{\hat{\sigma}^2}{2}\right)$$

This is called **Duan's smearing estimate** (in the general case) or the **log-normal retransformation**. Ignoring the $$e^{\sigma^2/2}$$ correction systematically **underestimates** the predicted $$y$$.

## Q20. You're told that adding a variable *increased* the RSS of your model. Is this possible in OLS? If not, why not? If yes, when?

**Answer:**

**No, this is IMPOSSIBLE in OLS with a standard model.**

Adding a variable to an OLS model can **never increase RSS** because:
- The larger model's parameter space *contains* the smaller model (set the new $$\beta_{p+1} = 0$$)
- OLS minimizes RSS over the parameter space
- A superset can only do as well or better than a subset: $$\text{RSS}_{p+1} \leq \text{RSS}_p$$

Formally: if $$S_1 \subset S_2$$, then $$\min_{\boldsymbol{\beta} \in S_2} \mathcal{L}(\boldsymbol{\beta}) \leq \min_{\boldsymbol{\beta} \in S_1} \mathcal{L}(\boldsymbol{\beta})$$.

**Exceptions** (where RSS *could* appear to increase):
- Using **regularized** regression (Ridge/LASSO) — the penalty can make RSS increase
- Numerical issues with near-singular matrices
- Comparing training RSS on *different* subsets of data
- Removing the intercept while adding a variable
