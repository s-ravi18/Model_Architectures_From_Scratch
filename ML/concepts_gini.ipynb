# Statistical & Machine Learning Concepts
### Detailed Mathematical Notes

## 1. Entropy vs Information Gain

### Recall: Entropy (Shannon Entropy)

**Definition:** Entropy quantifies the average uncertainty or impurity in a random variable. For a dataset $$S$$ with $$k$$ classes:

$$H(S) = -\sum_{i=1}^{k} p_i \log_2(p_i)$$

where $$p_i$$ is the proportion of class $$i$$ in $$S$$. By convention, $$0 \log_2 0 = 0$$.

**Range:** $$[0, \; \log_2(k)]$$
* $$H = 0$$: Pure node (all samples belong to one class)
* $$H = \log_2(k)$$: Maximum disorder (uniform distribution)
* Binary classification: $$[0, 1]$$

---

### Recall: Information Gain

**Definition:** Information Gain is the **reduction in entropy** achieved by partitioning dataset $$S$$ on attribute $$A$$:

$$IG(S, A) = H(S) - \sum_{v \in \text{Values}(A)} \frac{|S_v|}{|S|} \; H(S_v)$$

where $$S_v$$ is the subset of $$S$$ for which attribute $$A$$ takes value $$v$$.

**Range:** $$[0, \; H(S)]$$
* $$IG = 0$$: The split provides no new information
* $$IG = H(S)$$: The split perfectly separates classes

---

### Key Differences

| Aspect | Entropy | Information Gain |
| --- | --- | --- |
| Measures | Impurity of a **single** set | **Reduction** in impurity due to a split |
| Nature | Absolute measure of disorder | Differential (before − after) measure |
| Range | $$[0, \log_2 k]$$ | $$[0, H(S)]$$ |
| Usage | Quantifying disorder | Splitting criterion in ID3, C4.5 |

In short: entropy is the **state**, information gain is the **improvement**.

---

### Worked Example

Dataset $$S$$: 6 positive, 4 negative (10 total). Splitting on attribute $$A$$ yields:
* $$S_1$$: 5+, 1− (6 samples)
* $$S_2$$: 1+, 3− (4 samples)

$$H(S) = -\frac{6}{10}\log_2\frac{6}{10} - \frac{4}{10}\log_2\frac{4}{10} \approx 0.971$$

$$H(S_1) = -\frac{5}{6}\log_2\frac{5}{6} - \frac{1}{6}\log_2\frac{1}{6} \approx 0.650$$

$$H(S_2) = -\frac{1}{4}\log_2\frac{1}{4} - \frac{3}{4}\log_2\frac{3}{4} \approx 0.811$$

$$IG(S, A) = 0.971 - \frac{6}{10}(0.650) - \frac{4}{10}(0.811) = 0.971 - 0.390 - 0.324 = 0.257 \text{ bits}$$

The split reduces entropy by 0.257 bits — about 26% of the original uncertainty.

## 2. Information Gain vs Gini Impurity

### Recall: Gini Impurity

**Definition:** Gini impurity measures the probability that a randomly chosen sample would be misclassified if randomly labelled according to the class distribution:

$$G(S) = 1 - \sum_{i=1}^{k} p_i^2 = \sum_{i \neq j} p_i \, p_j$$

**Range:** $$[0, \; 1 - 1/k]$$
* $$G = 0$$: Pure node
* Binary classification: $$[0, \; 0.5]$$

**Gini Gain** (analogous to Information Gain):

$$\text{Gini Gain}(S, A) = G(S) - \sum_{v \in \text{Values}(A)} \frac{|S_v|}{|S|} \; G(S_v)$$

---

### Key Differences

| Aspect | Information Gain (Entropy-based) | Gini Impurity |
| --- | --- | --- |
| Formula | $$-\sum p_i \log_2 p_i$$ | $$1 - \sum p_i^2$$ |
| Binary range | $$[0, 1]$$ | $$[0, 0.5]$$ |
| Computation | Requires logarithm (slower) | Only squares (faster) |
| Sensitivity | More sensitive to changes in class distribution | Favours isolating the most frequent class |
| Bias | Biased toward multi-valued attributes (corrected by Gain Ratio) | Biased toward equal-sized partitions |
| Used in | ID3, C4.5 | CART |
| Peak (binary) | At $$p = 0.5$$: $$H = 1$$ | At $$p = 0.5$$: $$G = 0.5$$ |

### Geometric Comparison (Binary Case)

Both curves are concave and peak at $$p = 0.5$$, but Gini is a parabola $$2p(1-p)$$ while entropy is $$-p\log_2 p - (1-p)\log_2(1-p)$$. In practice, they rank candidate splits almost identically — disagreeing in fewer than 2% of cases.

---

### Example (Same Dataset as Q1)

$$G(S) = 1 - \left(\frac{6}{10}\right)^2 - \left(\frac{4}{10}\right)^2 = 1 - 0.36 - 0.16 = 0.48$$

$$G(S_1) = 1 - \left(\frac{5}{6}\right)^2 - \left(\frac{1}{6}\right)^2 \approx 0.278$$

$$G(S_2) = 1 - \left(\frac{1}{4}\right)^2 - \left(\frac{3}{4}\right)^2 = 0.375$$

$$\text{Gini Gain} = 0.48 - \frac{6}{10}(0.278) - \frac{4}{10}(0.375) = 0.48 - 0.167 - 0.150 = 0.163$$

Both metrics agree that the split is beneficial, though their numeric scales differ.

## 3. KL Divergence vs Cross Entropy

### Recall: Cross Entropy

**Definition:** Cross entropy measures the average number of bits needed to encode data from true distribution $$P$$ using a coding scheme optimised for approximate distribution $$Q$$:

$$H(P, Q) = -\sum_{x} P(x) \log Q(x)$$

**Range:** $$[H(P), \; +\infty)$$ — always ≥ the true entropy $$H(P)$$, with equality iff $$Q = P$$.

---

### Recall: KL Divergence (Kullback-Leibler Divergence)

**Definition:** KL divergence measures the **extra** bits needed when using $$Q$$ to encode data actually drawn from $$P$$:

$$D_{KL}(P \| Q) = \sum_{x} P(x) \log \frac{P(x)}{Q(x)} = \sum_{x} P(x) \log P(x) - \sum_{x} P(x) \log Q(x)$$

**Range:** $$[0, \; +\infty)$$
* $$D_{KL} = 0$$ iff $$P = Q$$
* **Not symmetric:** $$D_{KL}(P \| Q) \neq D_{KL}(Q \| P)$$ in general
* Not a true metric (violates symmetry and triangle inequality)

---

### The Fundamental Relationship

$$\boxed{H(P, Q) = H(P) + D_{KL}(P \| Q)}$$

$$\text{Cross Entropy} = \text{True Entropy} + \text{KL Divergence}$$

---

### Key Differences

| Aspect | Cross Entropy $$H(P, Q)$$ | KL Divergence $$D_{KL}(P \| Q)$$ |
| --- | --- | --- |
| Measures | **Total** bits to encode $$P$$ using $$Q$$ | **Extra** bits due to using $$Q$$ instead of $$P$$ |
| Minimum value | $$H(P)$$ (when $$Q = P$$) | $$0$$ (when $$Q = P$$) |
| Contains $$H(P)$$? | Yes | No (subtracts it out) |
| Symmetry | Not symmetric | Not symmetric |

---

### Why Cross Entropy Is Used as Loss (Not KL Divergence Directly)

In classification, $$P$$ is the true label distribution (one-hot, fixed). Since $$H(P)$$ is constant w.r.t. model parameters $$\theta$$:

$$\arg\min_\theta \; H(P, Q_\theta) = \arg\min_\theta \; D_{KL}(P \| Q_\theta)$$

Minimising cross entropy is **equivalent** to minimising KL divergence during training, but cross entropy is simpler to compute (no $$\log P(x)$$ term needed).

---

### Example

True distribution: $$P = [0.7, 0.2, 0.1]$$, Model prediction: $$Q = [0.5, 0.3, 0.2]$$

$$H(P) = -(0.7\ln 0.7 + 0.2\ln 0.2 + 0.1\ln 0.1) \approx 0.802 \text{ nats}$$

$$H(P, Q) = -(0.7\ln 0.5 + 0.2\ln 0.3 + 0.1\ln 0.2) \approx 0.968 \text{ nats}$$

$$D_{KL}(P \| Q) = H(P, Q) - H(P) = 0.968 - 0.802 = 0.166 \text{ nats}$$

The 0.166 nats represent the penalty for the model's imperfect approximation.

## 4. Probability of Predicting 1 in a Random Forest (Majority Voting)

### Setup

* $$n$$ trees in the random forest, each independently predicting class 1 with the same probability $$p$$
* Random forest uses **majority voting**: the sample is classified as 1 if more than half the trees predict 1

### Derivation

Each tree's prediction is a Bernoulli trial with parameter $$p$$. The total number of trees voting 1 follows a **Binomial distribution**:

$$K \sim \text{Binomial}(n, p)$$

The forest predicts 1 when $$K \geq \lceil n/2 \rceil$$ (assuming $$n$$ is odd for a clean majority):

$$\boxed{P(\text{RF predicts 1}) = \sum_{k=\lceil n/2 \rceil}^{n} \binom{n}{k} p^k (1-p)^{n-k}}$$

This can also be written using the **regularised incomplete beta function**:

$$P(\text{RF predicts 1}) = I_p\!\left(\lceil n/2 \rceil, \; n - \lceil n/2 \rceil + 1\right)$$

---

### Asymptotic Behaviour (Condorcet's Jury Theorem)

As $$n \to \infty$$:

$$P(\text{RF predicts 1}) \to \begin{cases} 1 & \text{if } p > 0.5 \\ 0.5 & \text{if } p = 0.5 \\ 0 & \text{if } p < 0.5 \end{cases}$$

**This is the power of ensembles:** if each individual tree is better than random ($$p > 0.5$$), the ensemble converges to a perfect classifier as $$n$$ grows. Conversely, if each tree is worse than random, ensembling makes things *worse*.

---

### Numerical Example

Let $$n = 5$$ trees, each with $$p = 0.6$$. Need $$\geq 3$$ trees to predict 1:

$$P = \sum_{k=3}^{5} \binom{5}{k} (0.6)^k (0.4)^{5-k}$$

$$= \binom{5}{3}(0.6)^3(0.4)^2 + \binom{5}{4}(0.6)^4(0.4)^1 + \binom{5}{5}(0.6)^5$$

$$= 10(0.216)(0.16) + 5(0.1296)(0.4) + 1(0.07776)$$

$$= 0.3456 + 0.2592 + 0.0778 = 0.6826$$

Each tree has 60% accuracy, but the 5-tree ensemble achieves **68.3%**.

With $$n = 101$$ trees at $$p = 0.6$$, the ensemble accuracy rises to \~97.8%. The improvement accelerates with more trees.

## 5. Variance Inflation Factor (VIF)

### Definition

VIF measures how much the variance of an estimated regression coefficient is **inflated** due to multicollinearity with other predictors. For the $$j$$-th predictor:

$$\boxed{\text{VIF}_j = \frac{1}{1 - R_j^2}}$$

where $$R_j^2$$ is the $$R^2$$ obtained by regressing $$X_j$$ on **all other** predictors $$X_1, \ldots, X_{j-1}, X_{j+1}, \ldots, X_p$$.

---

### Range and Interpretation

**Range:** $$[1, \; +\infty)$$

| $$\text{VIF}$$ | $$R_j^2$$ | Interpretation |
| --- | --- | --- |
| 1 | 0 | No collinearity — $$X_j$$ is uncorrelated with all others |
| 5 | 0.8 | Moderate collinearity — investigate |
| 10 | 0.9 | High collinearity — problematic |
| $$\to \infty$$ | $$\to 1$$ | Near-perfect collinearity |

---

### Connection to Coefficient Variance

The variance of the OLS estimator for $$\beta_j$$ can be decomposed as:

$$\text{Var}(\hat{\beta}_j) = \frac{\sigma^2}{\sum_{i=1}^{n}(x_{ij} - \bar{x}_j)^2} \cdot \text{VIF}_j = \frac{\sigma^2}{\text{SSX}_j} \cdot \text{VIF}_j$$

VIF directly **multiplies** the variance of $$\hat{\beta}_j$$. A VIF of 10 means the standard error of $$\hat{\beta}_j$$ is $$\sqrt{10} \approx 3.16$$ times larger than it would be without collinearity.

---

### How VIF Helps Solve Multicollinearity

1. **Detection:** Compute VIF for every predictor. High-VIF features are collinear.
2. **Diagnosis:** Regress each high-VIF feature on others to find *which* features are causing the inflation.
3. **Remediation** (guided by VIF values):
   * **Remove** one of the correlated features (higher VIF or less domain importance)
   * **Combine** correlated features (PCA, domain-driven composite scores)
   * **Regularise** (Ridge regression — see Q8)
   * **Increase sample size** (if collinearity is sample-driven)

---

### Example

| Feature | VIF | $$R_j^2$$ | Interpretation |
| --- | --- | --- | --- |
| $$X_1$$ (height_cm) | 45.2 | 0.978 | Severe — nearly explained by others |
| $$X_2$$ (height_in) | 44.8 | 0.978 | Severe — almost same info as $$X_1$$ |
| $$X_3$$ (weight) | 3.1 | 0.677 | Acceptable |
| $$X_4$$ (age) | 1.2 | 0.167 | No concern |

**Action:** Remove either $$X_1$$ or $$X_2$$ (they encode the same physical quantity in different units). Recompute VIFs after removal.

## 6. Why Adding Features Always Increases $$R^2$$

### Recall: $$R^2$$

$$R^2 = 1 - \frac{SS_{\text{res}}}{SS_{\text{tot}}} = 1 - \frac{\sum(y_i - \hat{y}_i)^2}{\sum(y_i - \bar{y})^2}$$

**Range:** $$[0, 1]$$ for OLS with an intercept.

---

### The Algebraic Proof

Consider two nested models:
* **Restricted model:** $$y = X_1\beta_1 + \epsilon$$ with $$p$$ features
* **Full model:** $$y = X_1\beta_1 + X_2\beta_2 + \epsilon$$ with $$p + q$$ features

**Key insight:** OLS minimises $$SS_{\text{res}}$$. The full model's parameter space is a **superset** of the restricted model's (set $$\beta_2 = 0$$ to recover the restricted model). The optimiser can always *at worst* replicate the restricted solution:

$$SS_{\text{res}}^{\text{full}} \leq SS_{\text{res}}^{\text{restricted}}$$

Since $$SS_{\text{tot}}$$ depends only on $$y$$ (fixed):

$$R^2_{\text{full}} = 1 - \frac{SS_{\text{res}}^{\text{full}}}{SS_{\text{tot}}} \geq 1 - \frac{SS_{\text{res}}^{\text{restricted}}}{SS_{\text{tot}}} = R^2_{\text{restricted}}$$

$$\boxed{R^2_{\text{full}} \geq R^2_{\text{restricted}}}$$

---

### Geometric Interpretation

OLS computes the **orthogonal projection** of $$y$$ onto the column space of $$X$$, denoted $$\text{Col}(X)$$. Adding features expands the column space:

$$\text{Col}(X_1) \subseteq \text{Col}([X_1, X_2])$$

Projecting onto a **larger** subspace always brings $$\hat{y}$$ at least as close to $$y$$. The residual length $$\|y - \hat{y}\|^2$$ can only shrink or stay the same — never grow.

---

### The Problem: Overfitting

Even **completely random noise features** increase $$R^2$$. With $$n$$ observations and $$p = n - 1$$ features, $$R^2 = 1$$ regardless of the data. This makes raw $$R^2$$ a misleading measure of model quality.

**Adjusted $$R^2$$** penalises model complexity:

$$R^2_{\text{adj}} = 1 - (1 - R^2)\frac{n - 1}{n - p - 1}$$

As $$p$$ increases, the penalty factor $$\frac{n-1}{n-p-1}$$ grows, so $$R^2_{\text{adj}}$$ **can decrease** if new features don't reduce $$SS_{\text{res}}$$ enough to justify the lost degrees of freedom.

## 7. Why Multicollinearity Causes Problems — Statistical Perspective

### The OLS Estimator and Its Variance

The OLS estimator:

$$\hat{\beta} = (X^TX)^{-1}X^Ty$$

Its variance-covariance matrix:

$$\text{Var}(\hat{\beta}) = \sigma^2 (X^TX)^{-1}$$

---

### What Happens Under Multicollinearity

When predictors are highly correlated, $$X^TX$$ becomes **near-singular**. Let $$\lambda_1 \geq \lambda_2 \geq \ldots \geq \lambda_p > 0$$ be its eigenvalues with corresponding eigenvectors $$v_1, \ldots, v_p$$.

The inverse decomposes as:

$$\text{Var}(\hat{\beta}) = \sigma^2 (X^TX)^{-1} = \sigma^2 \sum_{j=1}^{p} \frac{1}{\lambda_j} v_j v_j^T$$

**Near-collinearity** means $$\lambda_p \approx 0$$, so the term $$1/\lambda_p$$ **dominates**, inflating variance enormously in the direction of $$v_p$$.

---

### Consequences

**1. Inflated Coefficient Variance**

$$\text{Var}(\hat{\beta}_j) = \sigma^2 \cdot [(X^TX)^{-1}]_{jj} = \frac{\sigma^2}{\text{SSX}_j} \cdot \text{VIF}_j$$

High VIF → large variance → wide confidence intervals → coefficients are statistically unreliable.

**2. Unstable Coefficients**

Small changes in the data (adding/removing a few samples) cause **large swings** in $$\hat{\beta}$$. The condition number $$\kappa = \lambda_{\max} / \lambda_{\min}$$ quantifies this: high $$\kappa$$ means the solution is hypersensitive to perturbations.

**3. Unreliable Hypothesis Tests**

The t-statistic for testing $$H_0: \beta_j = 0$$:

$$t_j = \frac{\hat{\beta}_j}{\text{SE}(\hat{\beta}_j)}$$

When $$\text{SE}(\hat{\beta}_j)$$ is inflated, $$|t_j|$$ shrinks → we **fail to reject** $$H_0$$ even when $$\beta_j \neq 0$$. This causes **false negatives**: truly important features appear insignificant.

**4. Sign Instability**

The sign of $$\hat{\beta}_j$$ can flip across samples, making coefficient interpretation impossible (e.g., a coefficient appears negative when the true relationship is positive).

---

### What Multicollinearity Does NOT Affect

* **Predictions** $$\hat{y} = X\hat{\beta}$$ remain unbiased and stable on the training feature space
* **$$R^2$$** is not degraded
* **Unbiasedness** of $$\hat{\beta}$$ — OLS is still unbiased; the problem is **variance**, not bias

Multicollinearity specifically damages **individual coefficient estimates and their interpretability**, not the model's overall predictive power on in-sample data.

## 8. Which Regularisation for Multicollinear Features?

### Answer: **Ridge Regression (L2)** is the preferred choice.

---

### Recall: Ridge vs Lasso

| | Ridge (L2) | Lasso (L1) |
| --- | --- | --- |
| Penalty | $$\lambda \sum \beta_j^2$$ | $$\lambda \sum |\beta_j|$$ |
| Closed-form solution | $$\hat{\beta} = (X^TX + \lambda I)^{-1}X^Ty$$ | No closed form (requires iterative optimisation) |
| Feature selection | No — shrinks toward zero but never exactly zero | Yes — drives some $$\beta_j$$ to exactly 0 |
| Correlated features | Shrinks coefficients **evenly** across the group | Arbitrarily picks **one**, drops others |

---

### Why Ridge Works for Multicollinearity

**1. Fixes the Ill-Conditioned Matrix Directly**

The core problem is that $$X^TX$$ is near-singular (small eigenvalues). Ridge adds $$\lambda I$$:

$$(X^TX + \lambda I)$$

If $$X^TX$$ has eigenvalues $$\lambda_1, \ldots, \lambda_p$$, then $$(X^TX + \lambda I)$$ has eigenvalues $$\lambda_1 + \lambda, \ldots, \lambda_p + \lambda$$. The minimum eigenvalue is now **at least** $$\lambda > 0$$, making the matrix well-conditioned:

$$\kappa_{\text{ridge}} = \frac{\lambda_{\max} + \lambda}{\lambda_{\min} + \lambda} \ll \frac{\lambda_{\max}}{\lambda_{\min}} = \kappa_{\text{OLS}}$$

**2. Reduces Coefficient Variance**

In the eigenbasis of $$X^TX$$:

$$\text{Var}(\hat{\beta}_{\text{ridge},j}) = \sigma^2 \frac{\lambda_j}{(\lambda_j + \lambda)^2}$$

Compare with OLS: $$\text{Var}(\hat{\beta}_{\text{OLS},j}) = \sigma^2 / \lambda_j$$

For small $$\lambda_j$$ (the collinear directions), Ridge **dramatically** reduces variance.

**3. Bias-Variance Trade-off**

$$\text{MSE}(\hat{\beta}_{\text{ridge}}) = \text{Bias}^2 + \text{Variance}$$

Ridge introduces bias: $$\text{Bias}(\hat{\beta}_{\text{ridge}}) = -\lambda(X^TX + \lambda I)^{-1}\beta$$

But the **variance reduction** far outweighs the bias increase when multicollinearity is present, yielding a lower total MSE.

---

### Why Lasso Is Poor for Multicollinearity

Given a group of highly correlated features $$X_1 \approx X_2 \approx \ldots \approx X_m$$:

* **Lasso** selects **one** feature from the group and zeros the rest. *Which one* it selects is arbitrary and unstable — small data perturbations change the selection.
* **Ridge** distributes the coefficient magnitude **evenly** across the correlated group, producing stable, interpretable results.

---

### When to Use Elastic Net

If you need **both** collinearity handling **and** feature selection (sparse model), use **Elastic Net**:

$$\mathcal{L} = \|y - X\beta\|^2 + \lambda_1 \|\beta\|_1 + \lambda_2 \|\beta\|_2^2$$

Elastic Net inherits Ridge's stability for correlated groups (the **grouping effect**: correlated features are shrunk together) while retaining Lasso's ability to zero out truly irrelevant features.
